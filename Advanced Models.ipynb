```python
def check_date_range(df, stage_name, expected_days=None):
    # Đảm bảo cột Date đã ở định dạng datetime
    dates = pd.to_datetime(df['Date'])

    # Lấy ngày nhỏ nhất, lớn nhất và đếm số lượng ngày độc bản
    min_date = dates.min().strftime('%Y-%m-%d')
    max_date = dates.max().strftime('%Y-%m-%d')
    unique_days = dates.nunique()

    print(f"GIAI ĐOẠN: {stage_name.upper()}")
    print(f"Từ ngày:       {min_date}")
    print(f"Đến ngày:      {max_date}")
    print(f"Số ngày thực tế: {unique_days}")

    # Nếu có số ngày kỳ vọng, tiến hành kiểm tra tự động
    if expected_days is not None:
        if unique_days == expected_days:
            print("HỢP LỆ: Đủ số lượng ngày cấu hình.")
        else:
            print(f"LỖI: Thiếu hoặc thừa dữ liệu! Kỳ vọng {expected_days} ngày nhưng thực tế có {unique_days} ngày.")
    print("-" * 40)

# 1. Kiểm tra tập Train (Từ ngày đầu tiên đến hết 08/08/2025)
check_date_range(train_data_copy[train_data_copy['Date'] <= '2025-08-08'], "Train")

# 2. Kiểm tra tập Valid (Phải đủ 28 ngày: 09/08/2025 -> 05/09/2025)
valid_df = train_data_copy[(train_data_copy['Date'] >= '2025-08-09') & (train_data_copy['Date'] <= '2025-09-05')]
check_date_range(valid_df, "Validation", expected_days=28)
```

## Feature Engineering

```python
def build_forecasting_features(df_grouped, train_data_copy=None, date_col='Date',
                               item_col='ItemCode', target_col='TotalQuantity', qty_col='Quantity',
                               sales_amount_col='TotalSalesAmount',
                               cost_amount_col='TotalCostAmount',
                               unit_price_col='AverageUnitPrice',
                               unit_cost_col='AverageUnitCost',
                               val_size=28, alpha_return=30, use_static = False,
                               use_group = False):
    df = df_grouped.copy()
    df[date_col] = pd.to_datetime(df[date_col])
    df = df.sort_values([item_col, date_col]).reset_index(drop=True)

    actual_days_sorted = sorted(df[date_col].unique())
    train_days_list = actual_days_sorted[:-val_size]
    val_days_list = actual_days_sorted[-val_size:]

    # =========================================================
    # 1. STATIC RETURN RATE + STATIC SPARSITY
    # =========================================================
    if train_data_copy is not None and qty_col in train_data_copy.columns:
        tmp = train_data_copy.copy()
        tmp[date_col] = pd.to_datetime(tmp[date_col])
        tmp = tmp[tmp[date_col].isin(train_days_list)].copy()

        tmp['SalesQty_tmp'] = tmp[qty_col].clip(lower=0)
        tmp['ReturnQty_tmp'] = -tmp[qty_col].clip(upper=0)

        daily = (
            tmp.groupby([item_col, date_col], observed=True)
            .agg(
                sales_qty=('SalesQty_tmp', 'sum'),
                return_qty=('ReturnQty_tmp', 'sum')
            )
            .reset_index()
        )
    else:
        tmp = df[df[date_col].isin(train_days_list)].copy()
        tmp['sales_qty'] = tmp[target_col].clip(lower=0)
        tmp['return_qty'] = -tmp[target_col].clip(upper=0)

        daily = tmp[[item_col, date_col, 'sales_qty', 'return_qty']].copy()

    static_return = (
        daily.groupby(item_col, observed=True)
        .agg(
            sales_qty=('sales_qty', 'sum'),
            return_qty=('return_qty', 'sum'),
            n_days=(date_col, 'nunique'),
            active_days=('sales_qty', lambda x: (x > 0).sum())
        )
        .reset_index()
    )

    global_rate = (
        static_return['return_qty'].sum()
        / (static_return['sales_qty'].sum() + 1e-5)
    )

    static_return['static_return_rate'] = (
        static_return['return_qty'] + alpha_return * global_rate
    ) / (
        static_return['sales_qty'] + alpha_return
    )

    static_return['static_sparsity'] = 1 - (
        static_return['active_days']
        / static_return['n_days'].clip(lower=1)
    )

    static_return = static_return[
        [item_col, 'static_return_rate', 'static_sparsity']
    ]

    global_static_return_rate = static_return['static_return_rate'].mean()
    global_static_sparsity = static_return['static_sparsity'].mean()

    # =========================================================
    # 2. STATIC CV
    # =========================================================
    min_date = df[date_col].min()
    max_date = df[date_col].max()
    N = (max_date - min_date).days + 1

    static_cv = (
        df[df[date_col].isin(train_days_list)]
        .groupby(item_col, observed=True)
        .agg(
            sum_qty=(target_col, 'sum'),
            sum_sq_qty=(target_col, lambda x: np.sum(x ** 2))
        )
        .reset_index()
    )

    static_cv['avg_qty_real'] = static_cv['sum_qty'] / N

    variance_real = np.maximum(
        (static_cv['sum_sq_qty'] / N)
        - (static_cv['avg_qty_real'] ** 2),
        0
    )

    static_cv['std_qty_real'] = np.sqrt(variance_real)

    static_cv['static_cv'] = (
        static_cv['std_qty_real']
        / (static_cv['avg_qty_real'] + 1e-6)
    ).astype('float32')

    static_cv = static_cv[[item_col, 'static_cv']]
    global_static_cv = static_cv['static_cv'].mean()

    df = df.merge(static_cv, on=item_col, how='left')
    df = df.merge(static_return, on=item_col, how='left')

    df['static_cv'] = df['static_cv'].fillna(global_static_cv)
    df['static_sparsity'] = df['static_sparsity'].fillna(global_static_sparsity)
    df['static_return_rate'] = df['static_return_rate'].fillna(global_static_return_rate)

    # =========================================================
    # 3. BASIC FEATURES
    # =========================================================
    df[item_col] = df[item_col].astype('category')

    df['DailyProfit'] = df[sales_amount_col] - df[cost_amount_col]
    df['PriceCostRatio'] = df[unit_price_col] / (df[unit_cost_col] + 1e-5)
    df['ProfitMargin'] = df['DailyProfit'] / (df[sales_amount_col] + 1e-5)

    df['Weekday'] = df[date_col].dt.dayofweek
    df['Day'] = df[date_col].dt.day
    df['Month'] = df[date_col].dt.month
    df['IsWeekend'] = df['Weekday'].isin([5, 6]).astype(int)

    doy = df[date_col].dt.dayofyear
    df['sin_doy'] = np.sin(2 * np.pi * doy / 365.25).astype('float32')
    df['cos_doy'] = np.cos(2 * np.pi * doy / 365.25).astype('float32')

    # =========================================================
    # 4. LAG + ROLLING FEATURES
    # =========================================================
    g = df.groupby(item_col, observed=True)

    lag_config = {
        28: [28, 29, 35],
        56: [56, 57, 63]
    }

    for main_lag, lag_list in lag_config.items():
        for lag in lag_list:
            df[f'lag_{lag}'] = g[target_col].shift(lag)

        df[f'lag_{main_lag}_log'] = np.log1p(df[f'lag_{main_lag}'].fillna(0))

        df[f'rolling_mean_7_lag{main_lag}'] = (
            g[target_col]
            .transform(lambda x: x.shift(main_lag).rolling(7, min_periods=1).mean())
        )

        df[f'rolling_mean_14_lag{main_lag}'] = (
            g[target_col]
            .transform(lambda x: x.shift(main_lag).rolling(14, min_periods=1).mean())
        )

        df[f'rolling_mean_30_lag{main_lag}'] = (
            g[target_col]
            .transform(lambda x: x.shift(main_lag).rolling(30, min_periods=1).mean())
        )

        df[f'rolling_std_7_lag{main_lag}'] = (
            g[target_col]
            .transform(lambda x: x.shift(main_lag).rolling(7, min_periods=1).std())
        )

        df[f'rolling_std_14_lag{main_lag}'] = (
            g[target_col]
            .transform(lambda x: x.shift(main_lag).rolling(14, min_periods=1).std())
        )

        df[f'rolling_max_7_lag{main_lag}'] = (
            g[target_col]
            .transform(lambda x: x.shift(main_lag).rolling(7, min_periods=1).max())
        )

    # =========================================================
    # 5. DYNAMIC CV / SPARSITY / RETURN RATE LAG28 + LAG56
    # =========================================================
    df['SalesQty_tmp'] = df[target_col].clip(lower=0)
    df['ReturnQty_tmp'] = -df[target_col].clip(upper=0)
    df['QtySq_tmp'] = df['SalesQty_tmp'] ** 2
    df['IsZeroSale_tmp'] = (df['SalesQty_tmp'] == 0).astype(int)

    g = df.groupby(item_col, observed=True)

    df['cum_sales_qty_tmp'] = g['SalesQty_tmp'].cumsum()
    df['cum_return_qty_tmp'] = g['ReturnQty_tmp'].cumsum()
    df['cum_zero_days_tmp'] = g['IsZeroSale_tmp'].cumsum()
    df['cum_sum_qty_tmp'] = g['SalesQty_tmp'].cumsum()
    df['cum_sum_sq_qty_tmp'] = g['QtySq_tmp'].cumsum()
    df['cum_days_tmp'] = g.cumcount() + 1

    for k in [28, 56]:
        sales_lag = g['cum_sales_qty_tmp'].shift(k)
        return_lag = g['cum_return_qty_tmp'].shift(k)
        zero_days_lag = g['cum_zero_days_tmp'].shift(k)
        days_lag = g['cum_days_tmp'].shift(k)
        sum_qty_lag = g['cum_sum_qty_tmp'].shift(k)
        sum_sq_qty_lag = g['cum_sum_sq_qty_tmp'].shift(k)

        df[f'dynamic_return_rate_lag{k}'] = (
            (return_lag + 1.0) / (sales_lag + 20.0)
        )

        df[f'dynamic_sparsity_lag{k}'] = zero_days_lag / days_lag

        mean_qty = sum_qty_lag / days_lag
        variance_qty = (sum_sq_qty_lag / days_lag) - (mean_qty ** 2)
        variance_qty = np.maximum(variance_qty, 0)

        df[f'dynamic_cv_lag{k}'] = (
            np.sqrt(variance_qty) / (mean_qty + 1e-6)
        )

        df[f'dynamic_return_rate_lag{k}'] = (
            df[f'dynamic_return_rate_lag{k}']
            .replace([np.inf, -np.inf], np.nan)
            .fillna(1.0 / 20.0)
            .astype('float32')
        )

        df[f'dynamic_sparsity_lag{k}'] = (
            df[f'dynamic_sparsity_lag{k}']
            .replace([np.inf, -np.inf], np.nan)
            .fillna(1.0)
            .astype('float32')
        )

        df[f'dynamic_cv_lag{k}'] = (
            df[f'dynamic_cv_lag{k}']
            .replace([np.inf, -np.inf], np.nan)
            .fillna(0)
            .astype('float32')
        )

    tmp_cols = [c for c in df.columns if c.endswith('_tmp')]
    df = df.drop(columns=tmp_cols, errors='ignore')

    # =========================================================
    # 6. DAYS TO LAST SELL
    # =========================================================
    df['last_sell_date_tmp'] = np.where(
        df[target_col] > 0,
        df[date_col],
        pd.NaT
    )

    df['last_sell_date_tmp'] = pd.to_datetime(df['last_sell_date_tmp'])
    df['last_sell_date_tmp'] = (
        df.groupby(item_col, observed=True)['last_sell_date_tmp']
        .ffill()
    )

    df['days_to_last_sell'] = (
        df[date_col] - df['last_sell_date_tmp']
    ).dt.days

    df['days_to_last_sell'] = df['days_to_last_sell'].fillna(999)
    df = df.drop(columns=['last_sell_date_tmp'], errors='ignore')

    # =========================================================
    # 7. ISNA FLAGS + FILL LAG FEATURES
    # =========================================================
    lag_cols = [
        'lag_28', 'lag_29', 'lag_35',
        'lag_56', 'lag_57', 'lag_63',

        'rolling_mean_7_lag28',
        'rolling_mean_14_lag28',
        'rolling_mean_30_lag28',
        'rolling_std_7_lag28',
        'rolling_std_14_lag28',
        'rolling_max_7_lag28',

        'rolling_mean_7_lag56',
        'rolling_mean_14_lag56',
        'rolling_mean_30_lag56',
        'rolling_std_7_lag56',
        'rolling_std_14_lag56',
        'rolling_max_7_lag56'
    ]

    for col in lag_cols:
        df[f'{col}_isna'] = df[col].isna().astype(int)
        df[col] = df[col].fillna(0)

    # =========================================================
    # 8. WEIGHT WRMSSE
    # =========================================================
    sku_profit = (
        df.groupby(item_col, observed=True)['DailyProfit']
        .sum()
        .reset_index()
    )

    sku_profit['clean_profit'] = np.where(
        sku_profit['DailyProfit'] < 0,
        0,
        sku_profit['DailyProfit']
    )

    total_system_profit = sku_profit['clean_profit'].sum()

    sku_profit['weight_wrmsse'] = (
        sku_profit['clean_profit'] / (total_system_profit + 1e-5)
    )

    df = df.drop(columns=['weight_wrmsse'], errors='ignore')
    df = df.merge(
        sku_profit[[item_col, 'weight_wrmsse']],
        on=item_col,
        how='left'
    )

    # =========================================================
    # 9. SKU GROUP
    # =========================================================
    sku_group = (
        df[df[date_col].isin(train_days_list)]
        .groupby(item_col, observed=True)
        .agg(
            total_profit=('DailyProfit', 'sum'),
            total_qty=(target_col, 'sum'),
            active_days=(target_col, lambda x: (x > 0).sum()),
            n_days=(date_col, 'nunique')
        )
        .reset_index()
    )

    sku_group['positive_profit'] = sku_group['total_profit'].clip(lower=0)

    # Last selling date per SKU, train only
    last_sell = (
        df[
            (df[date_col].isin(train_days_list)) &
            (df[target_col] > 0)
        ]
        .groupby(item_col, observed=True)[date_col]
        .max()
        .reset_index()
        .rename(columns={date_col: 'last_sell_date'})
    )

    sku_group = sku_group.merge(last_sell, on=item_col, how='left')

    max_train_date = max(train_days_list)

    sku_group['days_since_last_sell'] = (
        max_train_date - sku_group['last_sell_date']
    ).dt.days

    sku_group['days_since_last_sell'] = (
        sku_group['days_since_last_sell'].fillna(9999)
    )

    sku_group['active_ratio'] = (
        sku_group['active_days'] / sku_group['n_days'].clip(lower=1)
    )

    def assign_sku_group(row):
        pp = row['positive_profit']

        if pp >= 28.28e6:
            return 'A'

        if pp >= 6.94e6:
            return 'B'

        if pp >= 2.58e6:
            return 'C'

        # Dead: SKU từng có profit nhưng lâu không bán
        if (pp > 0) and (row['days_since_last_sell'] >= 90):
            return 'Dead'

        # Slow: SKU bán rất thưa
        if (pp > 0) and (row['active_ratio'] <= 0.10):
            return 'Slow'

        # Medium: SKU còn lại, profit nhỏ
        if pp > 0:
            return 'Medium'

        # Profit <= 0: SKU không đóng góp positive profit
        return 'Profit <= 0'

    sku_group['sku_group'] = sku_group.apply(assign_sku_group, axis=1)

    df = df.merge(
        sku_group[
            [
                item_col,
                'positive_profit',
                'sku_group'
            ]
        ],
        on=item_col,
        how='left'
    )

    df['positive_profit'] = df['positive_profit'].fillna(0)
    df['sku_group'] = df['sku_group'].fillna('Profit <= 0')
    df['sku_group'] = df['sku_group'].astype('category')

    # =========================================================
    # 9. FEATURE LISTS
    # =========================================================
    if use_static:
      static_features = [
          'static_cv',
          'static_sparsity',
          'static_return_rate'
      ]
    else:
      static_features = []
    if use_group:
      group_features = [
          'positive_profit',
          'sku_group'
      ]
    else:
      group_features = []

    base_features_public = [
        item_col, 'Weekday', 'Day', 'Month', 'IsWeekend',
        'PriceCostRatio', 'days_to_last_sell', 'ProfitMargin',
        'sin_doy', 'cos_doy'
    ] + static_features + group_features + [
        'dynamic_cv_lag28',
        'dynamic_sparsity_lag28',
        'dynamic_return_rate_lag28'
    ]

    base_features_private = [
        item_col, 'Weekday', 'Day', 'Month', 'IsWeekend',
        'PriceCostRatio', 'days_to_last_sell', 'ProfitMargin',
        'sin_doy', 'cos_doy'
    ] + static_features + group_features + [
        'dynamic_cv_lag56',
        'dynamic_sparsity_lag56',
        'dynamic_return_rate_lag56'
    ]

    features_for_public = base_features_public + [
        'lag_28',
        'lag_29',
        'rolling_mean_7_lag28',
        'rolling_std_7_lag28',
        'rolling_max_7_lag28',
        'lag_28_log',
        'rolling_mean_14_lag28',
        'rolling_mean_30_lag28',
        'rolling_std_14_lag28',

        'lag_28_isna',
        'lag_29_isna',
        'rolling_mean_7_lag28_isna',
        'rolling_std_7_lag28_isna',
        'rolling_max_7_lag28_isna',
        'rolling_mean_14_lag28_isna',
        'rolling_mean_30_lag28_isna',
        'rolling_std_14_lag28_isna',
    ]

    features_for_private = base_features_private + [
        'lag_56',
        'lag_57',
        'rolling_mean_7_lag56',
        'rolling_std_7_lag56',
        'rolling_max_7_lag56',
        'lag_56_log',
        'rolling_mean_14_lag56',
        'rolling_mean_30_lag56',
        'rolling_std_14_lag56',

        'lag_56_isna',
        'lag_57_isna',
        'rolling_mean_7_lag56_isna',
        'rolling_std_7_lag56_isna',
        'rolling_max_7_lag56_isna',
        'rolling_mean_14_lag56_isna',
        'rolling_mean_30_lag56_isna',
        'rolling_std_14_lag56_isna',
    ]

    train_mask = df[date_col].isin(train_days_list)
    valid_mask = df[date_col].isin(val_days_list)

    output = {
        'df_daily': df,

        'features_for_public': features_for_public,
        'features_for_private': features_for_private,

        'X_train_public': df.loc[train_mask, features_for_public],
        'y_train_public': df.loc[train_mask, target_col],
        'X_val_public': df.loc[valid_mask, features_for_public],
        'y_val_public': df.loc[valid_mask, target_col],

        'X_train_private': df.loc[train_mask, features_for_private],
        'y_train_private': df.loc[train_mask, target_col],
        'X_val_private': df.loc[valid_mask, features_for_private],
        'y_val_private': df.loc[valid_mask, target_col],

        'train_days_list': train_days_list,
        'val_days_list': val_days_list
    }

    return output
```

## Model Class (Stacking LGBM + CAT -> RIDGE)

```python
import numpy as np
import pandas as pd

import lightgbm as lgb
from catboost import CatBoostRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import Ridge


class StackingTwinModelV2:
    """
    Kiến Trúc Lai Đa Tầng (Heavy Stacking Twin Model V2)
    ======================================================
    Tầng 1 - Base Models (4 mô hình):
        - LightGBM   : Boosting, độ chính xác cao, xử lý tốt dữ liệu thưa
        - CatBoost   : Boosting, mạnh với categorical features
        - RandomForest: Bagging, chống nhiễu khi dữ liệu 28 ngày sau biến động
        - Ridge L1   : Tuyến tính, bắt xu hướng tăng trưởng vượt đỉnh lịch sử

    Tầng 2 - Meta Model:
        - Ridge (positive=True): Kết hợp 4 pred từ Tầng 1,
          ràng buộc hệ số dương để tránh triệt tiêu tín hiệu

    Hai nhánh song song: Public & Private features
    """

    def __init__(
        self,
        features_public,
        features_private,
        target_col='TotalQuantity',
        date_col='Date',
        item_col='ItemCode',
        weight_col='weight_wrmsse',
        lgb_params=None,
        cat_params=None,
        rf_params=None,
        ridge_l1_alpha=1.0,
        ridge_l2_alpha=1.0,
    ):
        self.features_public = features_public
        self.features_private = features_private
        self.target_col = target_col
        self.date_col = date_col
        self.item_col = item_col
        self.weight_col = weight_col

        # ── LightGBM params ──────────────────────────────────────────────────
        self.lgb_params = lgb_params or {
            'objective': 'tweedie',
            'tweedie_variance_power': 1.15,
            'learning_rate': 0.05,
            'n_estimators': 500,
            'num_leaves': 63,
            'min_child_samples': 20,
            'verbosity': -1,
            'n_jobs': -1,
            'random_state': 42,
        }

        # ── CatBoost params ──────────────────────────────────────────────────
        self.cat_params = cat_params or {
            'loss_function': 'RMSE',
            'iterations': 500,
            'learning_rate': 0.05,
            'depth': 8,
            'random_seed': 42,
            'verbose': False,
            'task_type': 'CPU' # Changed from 'GPU' to 'CPU'
        }

        # ── RandomForest params ───────────────────────────────────────────────
        self.rf_params = rf_params or {
            'n_estimators': 300,
            'max_depth': 12,
            'min_samples_leaf': 20,
            'max_features': 'sqrt',
            'random_state': 42,
        }

        self.ridge_l1_alpha = ridge_l1_alpha
        self.ridge_l2_alpha = ridge_l2_alpha

        # ── Khởi tạo models ───────────────────────────────────────────────────
        self._init_models()

    # =========================================================================
    #  Khởi tạo / Reset models
    # =========================================================================
    def _init_models(self):
        # Base models – Public
        self.model_lgb_public = lgb.LGBMRegressor(**self.lgb_params)
        self.model_cat_public = CatBoostRegressor(**self.cat_params)
        self.model_rf_public  = RandomForestRegressor(**self.rf_params)
        self.model_ridge_l1_public = Ridge(alpha=self.ridge_l1_alpha)

        # Base models – Private
        self.model_lgb_private = lgb.LGBMRegressor(**self.lgb_params)
        self.model_cat_private = CatBoostRegressor(**self.cat_params)
        self.model_rf_private  = RandomForestRegressor(**self.rf_params)
        self.model_ridge_l1_private = Ridge(alpha=self.ridge_l1_alpha)

        # Meta models Tầng 2 (positive=True để ràng buộc hệ số không âm)
        self.meta_public  = Ridge(alpha=self.ridge_l2_alpha, positive=True)
        self.meta_private = Ridge(alpha=self.ridge_l2_alpha, positive=True)

    # =========================================================================
    #  WRMSSE helpers
    # =========================================================================
    def _prepare_wrmsse_weight(self, df_history):
        df_history = df_history.sort_values(
            [self.item_col, self.date_col]
        ).copy()

        df_history['diff_sq'] = (
            df_history
            .groupby(self.item_col)[self.target_col]
            .diff()
            .pow(2)
        )

        scale_df = (
            df_history
            .groupby(self.item_col)['diff_sq']
            .mean()
            .reset_index()
            .rename(columns={'diff_sq': 'scale_denom'})
        )
        scale_df['scale_denom'] = (
            scale_df['scale_denom'].replace(0, 1e-5).fillna(1e-5)
        )

        weight_df = (
            df_history[[self.item_col, self.weight_col]].drop_duplicates()
        )

        wrmsse_df = weight_df.merge(scale_df, on=self.item_col, how='left')
        wrmsse_df['scale_denom'] = wrmsse_df['scale_denom'].fillna(1e-5)
        wrmsse_df['wrmsse_train_weight'] = (
            wrmsse_df[self.weight_col] / wrmsse_df['scale_denom']
        )
        return wrmsse_df[[self.item_col, 'scale_denom', 'wrmsse_train_weight']]

    def _add_wrmsse_weight_to_df(self, df, wrmsse_df):
        df_out = df.merge(wrmsse_df, on=self.item_col, how='left')
        df_out['wrmsse_train_weight'] = (
            df_out['wrmsse_train_weight'].fillna(0)
        )
        return df_out

    def _calculate_wrmsse(self, df_val, preds, df_train_history):
        df_train_history = df_train_history.sort_values(
            [self.item_col, self.date_col]
        ).copy()

        df_train_history['diff_sq'] = (
            df_train_history
            .groupby(self.item_col)[self.target_col]
            .diff()
            .pow(2)
        )

        scale_denom = (
            df_train_history
            .groupby(self.item_col)['diff_sq']
            .mean()
            .reset_index()
            .rename(columns={'diff_sq': 'denominator'})
        )
        scale_denom['denominator'] = (
            scale_denom['denominator'].replace(0, 1e-5).fillna(1e-5)
        )

        item_weights = (
            df_train_history[[self.item_col, self.weight_col]]
            .drop_duplicates()
        )

        df_eval = df_val[[self.item_col, self.target_col]].copy()
        df_eval['pred'] = np.clip(preds, 0, None)
        df_eval['se']   = (df_eval[self.target_col] - df_eval['pred']) ** 2

        val_mse = (
            df_eval
            .groupby(self.item_col)['se']
            .mean()
            .reset_index()
            .rename(columns={'se': 'mse'})
        )

        metric_df = val_mse.merge(scale_denom, on=self.item_col, how='left')
        metric_df['denominator'] = metric_df['denominator'].fillna(1e-5)
        metric_df['rmsse'] = np.sqrt(
            metric_df['mse'] / metric_df['denominator']
        )

        metric_df = metric_df.merge(item_weights, on=self.item_col, how='left')
        metric_df[self.weight_col] = metric_df[self.weight_col].fillna(0)

        return np.sum(metric_df['rmsse'] * metric_df[self.weight_col])

    # =========================================================================
    #  Encode string/category columns → int cho sklearn (RF, Ridge)
    #  LightGBM và CatBoost tự xử lý được nên KHÔNG cần encode.
    # =========================================================================
    def _encode_for_sklearn(self, X: pd.DataFrame) -> pd.DataFrame:
        """
        Chuyển tất cả cột object/category sang integer codes.
        Dùng riêng cho RandomForest và Ridge (sklearn không nhận string).
        """
        X_enc = X.copy()
        for col in X_enc.columns:
            if X_enc[col].dtype == 'object':
                X_enc[col] = X_enc[col].astype('category').cat.codes
            elif hasattr(X_enc[col], 'cat'):          # đã là category dtype
                X_enc[col] = X_enc[col].cat.codes
        return X_enc

    # =========================================================================
    #  Internal: train 4 base models → trả về OOF predictions
    # =========================================================================
    def _train_base_models(
        self,
        branch,          # 'public' | 'private'
        features,
        X_train, y_train, w_train,
        X_val,   y_val,  w_val,
        cat_features,
    ):
        lgb_model   = getattr(self, f'model_lgb_{branch}')
        cat_model   = getattr(self, f'model_cat_{branch}')
        rf_model    = getattr(self, f'model_rf_{branch}')
        ridge_model = getattr(self, f'model_ridge_l1_{branch}')

        # LightGBM & CatBoost xử lý được string/category trực tiếp
        # RandomForest & Ridge (sklearn) cần encode sang int trước
        X_train_enc = self._encode_for_sklearn(X_train)
        X_val_enc   = self._encode_for_sklearn(X_val)

        print(f"  [{branch.capitalize()}] Train LightGBM ...")
        lgb_model.fit(X_train, y_train, sample_weight=w_train)

        print(f"  [{branch.capitalize()}] Train CatBoost ...")
        cat_model.fit(
            X_train, y_train,
            sample_weight=w_train,
            cat_features=cat_features,
        )

        print(f"  [{branch.capitalize()}] Train RandomForest ...")
        rf_model.fit(X_train_enc, y_train, sample_weight=w_train)

        print(f"  [{branch.capitalize()}] Train Ridge L1 ...")
        ridge_model.fit(X_train_enc, y_train, sample_weight=w_train)

        # OOF preds trên val
        pred_lgb   = lgb_model.predict(X_val)
        pred_cat   = cat_model.predict(X_val)
        pred_rf    = rf_model.predict(X_val_enc)
        pred_ridge = ridge_model.predict(X_val_enc)

        return pred_lgb, pred_cat, pred_rf, pred_ridge

    def _get_cat_feature_indices(self, features):
        cat_cols = [
            self.item_col,
            'sku_group'
        ]
        return [
            i for i, col in enumerate(features)
            if col in cat_cols
        ]

    # =========================================================================
    #  GIAI ĐOẠN 1: Train base + stacker trên train/val split
    # =========================================================================
    def fit_giai_doan_1(self, df, train_days, val_days):
        print("=" * 65)
        print("GIAI ĐOẠN 1: TRAIN BASE MODELS (4) + TRAIN RIDGE STACKER L2")
        print("=" * 65)

        mask_train = df[self.date_col].isin(train_days)
        mask_val   = df[self.date_col].isin(val_days)

        df_train_raw = df.loc[mask_train].copy()
        df_val_raw   = df.loc[mask_val].copy()

        wrmsse_df = self._prepare_wrmsse_weight(df_train_raw)
        df_train  = self._add_wrmsse_weight_to_df(df_train_raw, wrmsse_df)
        df_val    = self._add_wrmsse_weight_to_df(df_val_raw,   wrmsse_df)

        y_train = df_train[self.target_col].clip(lower=0)
        y_val   = df_val[self.target_col].clip(lower=0)
        w_train = df_train['wrmsse_train_weight'].values
        w_val   = df_val['wrmsse_train_weight'].values

        results = {}

        # ── PUBLIC ──────────────────────────────────────────────────────────
        print("\n[PUBLIC BRANCH]")
        cat_idx_pub = self._get_cat_feature_indices(self.features_public)

        pred_lgb_pub, pred_cat_pub, pred_rf_pub, pred_ridge_pub = (
            self._train_base_models(
                branch='public',
                features=self.features_public,
                X_train=df_train[self.features_public],
                y_train=y_train, w_train=w_train,
                X_val=df_val[self.features_public],
                y_val=y_val, w_val=w_val,
                cat_features=cat_idx_pub,
            )
        )

        # Tầng 2: Ridge (positive=True) fit trên 4 OOF preds
        X_meta_pub = np.column_stack([
            pred_lgb_pub,
            pred_cat_pub,
            pred_rf_pub,
            pred_ridge_pub,
        ])
        self.meta_public.fit(X_meta_pub, y_val, sample_weight=w_val)

        final_pub_val = self.meta_public.predict(X_meta_pub)
        wrmsse_pub = self._calculate_wrmsse(
            df_val_raw, final_pub_val, df_train_raw
        )
        print(f"  [Public]  WRMSSE Validation : {wrmsse_pub:.6f}")
        print(f"  [Public]  Ridge L2 coefs    : {self.meta_public.coef_}")

        results['wrmsse_public'] = wrmsse_pub

        # ── PRIVATE ─────────────────────────────────────────────────────────
        print("\n[PRIVATE BRANCH]")
        cat_idx_pri = self._get_cat_feature_indices(self.features_private)

        pred_lgb_pri, pred_cat_pri, pred_rf_pri, pred_ridge_pri = (
            self._train_base_models(
                branch='private',
                features=self.features_private,
                X_train=df_train[self.features_private],
                y_train=y_train, w_train=w_train,
                X_val=df_val[self.features_private],
                y_val=y_val, w_val=w_val,
                cat_features=cat_idx_pri,
            )
        )

        X_meta_pri = np.column_stack([
            pred_lgb_pri,
            pred_cat_pri,
            pred_rf_pri,
            pred_ridge_pri,
        ])
        self.meta_private.fit(X_meta_pri, y_val, sample_weight=w_val)

        final_pri_val = self.meta_private.predict(X_meta_pri)
        wrmsse_pri = self._calculate_wrmsse(
            df_val_raw, final_pri_val, df_train_raw
        )
        print(f"  [Private] WRMSSE Validation : {wrmsse_pri:.6f}")
        print(f"  [Private] Ridge L2 coefs    : {self.meta_private.coef_}")

        results['wrmsse_private'] = wrmsse_pri
        return results

    # =========================================================================
    #  GIAI ĐOẠN 2: Retrain toàn bộ base models trên full history
    #               (Ridge stacker L2 giữ nguyên hệ số từ val)
    # =========================================================================
    def retrain_giai_doan_2(self, df, history_days):
        print("=" * 65)
        print("GIAI ĐOẠN 2: RETRAIN 4 BASE MODELS TRÊN TOÀN BỘ HISTORY")
        print("           (Ridge Stacker L2 giữ nguyên từ validation)")
        print("=" * 65)

        mask_history = df[self.date_col].isin(history_days)
        df_history_raw = df.loc[mask_history].copy()

        wrmsse_df  = self._prepare_wrmsse_weight(df_history_raw)
        df_history = self._add_wrmsse_weight_to_df(df_history_raw, wrmsse_df)

        y = df_history[self.target_col].clip(lower=0)
        w = df_history['wrmsse_train_weight'].values

        cat_idx_pub = self._get_cat_feature_indices(self.features_public)
        cat_idx_pri = self._get_cat_feature_indices(self.features_private)

        for branch, features, cat_idx in [
            ('public',  self.features_public,  cat_idx_pub),
            ('private', self.features_private, cat_idx_pri),
        ]:
            X     = df_history[features]
            X_enc = self._encode_for_sklearn(X)
            print(f"\n  Retrain [{branch.capitalize()}] base models ...")

            getattr(self, f'model_lgb_{branch}').fit(X, y, sample_weight=w)

            getattr(self, f'model_cat_{branch}').fit(
                X, y, sample_weight=w, cat_features=cat_idx
            )

            getattr(self, f'model_rf_{branch}').fit(X_enc, y, sample_weight=w)

            getattr(self, f'model_ridge_l1_{branch}').fit(
                X_enc, y, sample_weight=w
            )

        print("\nRetrain 4 base models hoàn tất.")
        print("Ridge Stacker L2 giữ nguyên hệ số từ giai đoạn validation.")

    # =========================================================================
    #  Predict submission
    # =========================================================================
    def predict_submission(self, df_test_public, df_test_private):
        print("Đang predict submission với 4+1 layer stacking ...")

        def _predict_branch(branch, features, df_test):
            X_raw = df_test[features]
            X_enc = self._encode_for_sklearn(X_raw)
            pred_lgb   = getattr(self, f'model_lgb_{branch}').predict(X_raw)
            pred_cat   = getattr(self, f'model_cat_{branch}').predict(X_raw)
            pred_rf    = getattr(self, f'model_rf_{branch}').predict(X_enc)
            pred_ridge = getattr(self, f'model_ridge_l1_{branch}').predict(X_enc)

            X_meta = np.column_stack([
                pred_lgb, pred_cat, pred_rf, pred_ridge
            ])
            final = getattr(self, f'meta_{branch}').predict(X_meta)
            return np.clip(final, 0, None)

        final_pub = _predict_branch(
            'public', self.features_public, df_test_public
        )
        final_pri = _predict_branch(
            'private', self.features_private, df_test_private
        )

        return final_pub, final_pri
```

```python
import numpy as np
import pandas as pd


class ForecastSubmissionInferencer:
    def __init__(self, stack_model, features_for_public, features_for_private,
                 date_col='Date', item_col='ItemCode', target_col='TotalQuantity',
                 horizon_total=56, horizon_each=28):
        self.stack_model = stack_model
        self.features_for_public = features_for_public
        self.features_for_private = features_for_private
        self.date_col = date_col
        self.item_col = item_col
        self.target_col = target_col
        self.horizon_total = horizon_total
        self.horizon_each = horizon_each

    def _get_clean_skus(self, sample_submission):
        id_col = 'id' if 'id' in sample_submission.columns else self.item_col

        sample_submission = sample_submission.copy()
        sample_submission['clean_sku'] = (
            sample_submission[id_col]
            .astype(str)
            .str.replace('_validation', '', regex=False)
            .str.replace('_evaluation', '', regex=False)
        )

        return sample_submission['clean_sku'].unique()

    def _make_future_frame(self, df_daily, sample_submission):
        unique_skus = self._get_clean_skus(sample_submission)

        max_train_date = pd.to_datetime(df_daily[self.date_col].max())

        future_dates = [
            max_train_date + pd.to_timedelta(i, unit='D')
            for i in range(1, self.horizon_total + 1)
        ]

        df_future = (
            pd.MultiIndex
            .from_product(
                [unique_skus, future_dates],
                names=[self.item_col, self.date_col]
            )
            .to_frame(index=False)
        )

        df_future[self.target_col] = np.nan
        df_future['is_test'] = 1

        return df_future

    def _add_calendar_features(self, df):
        df[self.date_col] = pd.to_datetime(df[self.date_col])

        df['Weekday'] = df[self.date_col].dt.weekday
        df['Day'] = df[self.date_col].dt.day
        df['Month'] = df[self.date_col].dt.month
        df['IsWeekend'] = (df['Weekday'] >= 5).astype(int)

        doy = df[self.date_col].dt.dayofyear
        df['sin_doy'] = np.sin(2 * np.pi * doy / 365.25).astype('float32')
        df['cos_doy'] = np.cos(2 * np.pi * doy / 365.25).astype('float32')

        return df

    def _add_lag_rolling_features(self, df):
        g = df.groupby(self.item_col, observed=True)

        lag_config = {
            28: [28, 29, 35],
            56: [56, 57, 63]
        }

        for main_lag, lag_list in lag_config.items():
            for lag in lag_list:
                df[f'lag_{lag}'] = g[self.target_col].shift(lag)

            df[f'lag_{main_lag}_log'] = np.log1p(
                df[f'lag_{main_lag}'].fillna(0)
            )

            df[f'rolling_mean_7_lag{main_lag}'] = (
                g[self.target_col]
                .transform(lambda x: x.shift(main_lag).rolling(7, min_periods=1).mean())
            )

            df[f'rolling_mean_14_lag{main_lag}'] = (
                g[self.target_col]
                .transform(lambda x: x.shift(main_lag).rolling(14, min_periods=1).mean())
            )

            df[f'rolling_mean_30_lag{main_lag}'] = (
                g[self.target_col]
                .transform(lambda x: x.shift(main_lag).rolling(30, min_periods=1).mean())
            )

            df[f'rolling_std_7_lag{main_lag}'] = (
                g[self.target_col]
                .transform(lambda x: x.shift(main_lag).rolling(7, min_periods=1).std())
            )

            df[f'rolling_std_14_lag{main_lag}'] = (
                g[self.target_col]
                .transform(lambda x: x.shift(main_lag).rolling(14, min_periods=1).std())
            )

            df[f'rolling_max_7_lag{main_lag}'] = (
                g[self.target_col]
                .transform(lambda x: x.shift(main_lag).rolling(7, min_periods=1).max())
            )

        lag_cols = [
            'lag_28', 'lag_29', 'lag_35',
            'lag_56', 'lag_57', 'lag_63',

            'rolling_mean_7_lag28',
            'rolling_mean_14_lag28',
            'rolling_mean_30_lag28',
            'rolling_std_7_lag28',
            'rolling_std_14_lag28',
            'rolling_max_7_lag28',

            'rolling_mean_7_lag56',
            'rolling_mean_14_lag56',
            'rolling_mean_30_lag56',
            'rolling_std_7_lag56',
            'rolling_std_14_lag56',
            'rolling_max_7_lag56'
        ]

        for col in lag_cols:
            df[f'{col}_isna'] = df[col].isna().astype(int)
            df[col] = df[col].fillna(0)

        return df

    def _add_dynamic_features(self, df):
        g = df.groupby(self.item_col, observed=True)

        df['SalesQty_tmp'] = df[self.target_col].clip(lower=0)
        df['ReturnQty_tmp'] = -df[self.target_col].clip(upper=0)
        df['QtySq_tmp'] = df['SalesQty_tmp'] ** 2
        df['IsZeroSale_tmp'] = (df['SalesQty_tmp'] == 0).astype(int)

        df['cum_sales_qty_tmp'] = g['SalesQty_tmp'].cumsum()
        df['cum_return_qty_tmp'] = g['ReturnQty_tmp'].cumsum()
        df['cum_zero_days_tmp'] = g['IsZeroSale_tmp'].cumsum()
        df['cum_sum_qty_tmp'] = g['SalesQty_tmp'].cumsum()
        df['cum_sum_sq_qty_tmp'] = g['QtySq_tmp'].cumsum()
        df['cum_days_tmp'] = g.cumcount() + 1

        g = df.groupby(self.item_col, observed=True)

        for k in [28, 56]:
            sales_lag = g['cum_sales_qty_tmp'].shift(k)
            return_lag = g['cum_return_qty_tmp'].shift(k)
            zero_days_lag = g['cum_zero_days_tmp'].shift(k)
            days_lag = g['cum_days_tmp'].shift(k)
            sum_qty_lag = g['cum_sum_qty_tmp'].shift(k)
            sum_sq_qty_lag = g['cum_sum_sq_qty_tmp'].shift(k)

            df[f'dynamic_return_rate_lag{k}'] = (
                (return_lag + 1.0) / (sales_lag + 20.0)
            )

            df[f'dynamic_sparsity_lag{k}'] = zero_days_lag / days_lag

            mean_qty = sum_qty_lag / days_lag
            variance_qty = (sum_sq_qty_lag / days_lag) - (mean_qty ** 2)
            variance_qty = np.maximum(variance_qty, 0)

            df[f'dynamic_cv_lag{k}'] = (
                np.sqrt(variance_qty) / (mean_qty + 1e-6)
            )

            df[f'dynamic_return_rate_lag{k}'] = (
                df[f'dynamic_return_rate_lag{k}']
                .replace([np.inf, -np.inf], np.nan)
                .fillna(1.0 / 20.0)
                .astype('float32')
            )

            df[f'dynamic_sparsity_lag{k}'] = (
                df[f'dynamic_sparsity_lag{k}']
                .replace([np.inf, -np.inf], np.nan)
                .fillna(1.0)
                .astype('float32')
            )

            df[f'dynamic_cv_lag{k}'] = (
                df[f'dynamic_cv_lag{k}']
                .replace([np.inf, -np.inf], np.nan)
                .fillna(0)
                .astype('float32')
            )

        tmp_cols = [c for c in df.columns if c.endswith('_tmp')]
        df = df.drop(columns=tmp_cols, errors='ignore')

        return df

    def _forward_fill_static_features(self, df):
        ffill_cols = [
            'PriceCostRatio',
            'ProfitMargin',
            'static_cv',
            'static_sparsity',
            'static_return_rate',
            'weight_wrmsse',
            'positive_profit'
        ]

        for col in ffill_cols:
            if col in df.columns:
                df[col] = df.groupby(self.item_col, observed=True)[col].ffill()

        if 'positive_profit' in df.columns:
            df['positive_profit'] = df['positive_profit'].fillna(0)

        if 'sku_group' in df.columns:
            df['sku_group'] = df['sku_group'].fillna('Profit <= 0')
            df['sku_group'] = df['sku_group'].astype('category')

        return df

    def _update_days_to_last_sell(self, df):
        if 'days_to_last_sell' not in df.columns:
            return df

        df['last_sell_date_calculated'] = np.where(
            df['is_test'] == 0,
            df[self.date_col] - pd.to_timedelta(df['days_to_last_sell'], unit='D'),
            pd.NaT
        )

        df['last_sell_date_calculated'] = pd.to_datetime(
            df['last_sell_date_calculated']
        )

        df['last_sell_date_calculated'] = (
            df.groupby(self.item_col, observed=True)['last_sell_date_calculated']
            .ffill()
        )

        df['days_to_last_sell'] = (
            df[self.date_col] - df['last_sell_date_calculated']
        ).dt.days

        df['days_to_last_sell'] = df['days_to_last_sell'].fillna(999)

        df = df.drop(columns=['last_sell_date_calculated'], errors='ignore')

        return df

    def _sync_lgbm_categories(self, df_test_all):
        booster = self.stack_model.model_lgb_public.booster_
        feature_names = booster.feature_name()
        pandas_categorical = booster.pandas_categorical

        cat_features = [
            self.item_col,
            'sku_group'
        ]

        for col in cat_features:
            if col not in df_test_all.columns:
                continue

            if col not in feature_names:
                continue

            idx = feature_names.index(col)

            if idx >= len(pandas_categorical):
                continue

            categories = pandas_categorical[idx]

            df_test_all[col] = pd.Categorical(
                df_test_all[col].astype(str),
                categories=categories
            )

        return df_test_all

    def prepare_master(self, df_daily, sample_submission):
        df_hist = df_daily.copy()
        df_future = self._make_future_frame(df_hist, sample_submission)

        df_hist['is_test'] = 0

        df_hist[self.item_col] = df_hist[self.item_col].astype(str)
        df_future[self.item_col] = df_future[self.item_col].astype(str)

        df_master = pd.concat([df_hist, df_future], axis=0).reset_index(drop=True)

        df_master = (
            df_master
            .sort_values([self.item_col, self.date_col])
            .reset_index(drop=True)
        )

        df_master = self._add_calendar_features(df_master)
        df_master = self._forward_fill_static_features(df_master)
        if 'sku_group' in df_master.columns:
            df_master['sku_group'] = (
                df_master['sku_group']
                .fillna('Profit <= 0')
                .astype('category')
            )

        if 'positive_profit' in df_master.columns:
            df_master['positive_profit'] = (
                df_master['positive_profit']
                .fillna(0)
            )
        df_master = self._update_days_to_last_sell(df_master)
        df_master = self._add_lag_rolling_features(df_master)
        df_master = self._add_dynamic_features(df_master)

        df_master[self.item_col] = df_master[self.item_col].astype('category')

        return df_master

    def predict_test(self, df_master):
        df_test_all = df_master[df_master['is_test'] == 1].copy()

        df_test_all[self.item_col] = df_test_all[self.item_col].astype(str)
        df_test_all = self._sync_lgbm_categories(df_test_all)

        unique_test_days = sorted(df_test_all[self.date_col].unique())

        assert len(unique_test_days) == self.horizon_total

        public_days = unique_test_days[:self.horizon_each]
        private_days = unique_test_days[self.horizon_each:]

        df_test_public = df_test_all[
            df_test_all[self.date_col].isin(public_days)
        ].copy()

        df_test_private = df_test_all[
            df_test_all[self.date_col].isin(private_days)
        ].copy()

        missing_pub = set(self.features_for_public) - set(df_test_public.columns)
        missing_pri = set(self.features_for_private) - set(df_test_private.columns)

        if missing_pub:
            raise ValueError(f"Missing public features: {missing_pub}")

        if missing_pri:
            raise ValueError(f"Missing private features: {missing_pri}")

        public_preds, private_preds = self.stack_model.predict_submission(
            df_test_public=df_test_public,
            df_test_private=df_test_private
        )

        df_test_public[self.target_col] = public_preds
        df_test_private[self.target_col] = private_preds

        return df_test_public, df_test_private

    def _pivot_to_wide_format(self, df_partition, suffix):
        df_partition = df_partition.copy()
        df_partition[self.item_col] = df_partition[self.item_col].astype(str)

        df_partition = df_partition.sort_values([self.item_col, self.date_col])

        df_partition['day_idx'] = (
            df_partition.groupby(self.item_col).cumcount() + 1
        )

        df_partition['F_col'] = 'F' + df_partition['day_idx'].astype(str)

        df_pivot = df_partition.pivot(
            index=self.item_col,
            columns='F_col',
            values=self.target_col
        )

        f_cols = [f'F{i}' for i in range(1, self.horizon_each + 1)]

        df_pivot = df_pivot.reindex(columns=f_cols).reset_index()
        df_pivot['id'] = df_pivot[self.item_col].astype(str) + suffix

        return df_pivot[['id'] + f_cols]

    def make_submission(self, df_test_public, df_test_private, sample_submission):
        f_cols = [f'F{i}' for i in range(1, self.horizon_each + 1)]

        df_public_wide = self._pivot_to_wide_format(
            df_test_public,
            suffix='_validation'
        )

        df_private_wide = self._pivot_to_wide_format(
            df_test_private,
            suffix='_evaluation'
        )

        df_preds = pd.concat(
            [df_public_wide, df_private_wide],
            axis=0
        ).reset_index(drop=True)

        df_final = sample_submission[['id']].merge(
            df_preds,
            on='id',
            how='left'
        )

        df_final[f_cols] = df_final[f_cols].fillna(0)
        df_final[f_cols] = df_final[f_cols].clip(lower=0)

        assert df_final.shape[0] == sample_submission.shape[0]
        assert df_final[f_cols].isna().sum().sum() == 0

        return df_final

    def run(self, df_daily, sample_submission, output_path=None):
        sample_submission = sample_submission.copy()

        df_master = self.prepare_master(df_daily, sample_submission)

        df_test_public, df_test_private = self.predict_test(df_master)

        df_submission = self.make_submission(
            df_test_public,
            df_test_private,
            sample_submission
        )

        if output_path is not None:
            df_submission.to_csv(output_path, index=False)

        return {
            'df_master': df_master,
            'df_test_public': df_test_public,
            'df_test_private': df_test_private,
            'submission': df_submission
        }
```

# Recursive Forecasting

## Feature Engineering

```python
import numpy as np
import pandas as pd


def build_recursive_train_features(
    df_grouped,
    date_col='Date',
    item_col='ItemCode',
    target_col='TotalQuantity',
    sales_amount_col='TotalSalesAmount',
    cost_amount_col='TotalCostAmount',
    unit_price_col='AverageUnitPrice',
    unit_cost_col='AverageUnitCost',
    lags=(1, 2, 7, 14, 28),
    rolling_windows=(7, 14, 28),
    val_size=28,
    use_static=True
):
    df = df_grouped.copy()

    df[date_col] = pd.to_datetime(df[date_col])
    df = df.sort_values([item_col, date_col]).reset_index(drop=True)

    actual_days_sorted = sorted(df[date_col].unique())
    train_days_list = actual_days_sorted[:-val_size]
    val_days_list = actual_days_sorted[-val_size:]

    # =====================================================
    # 1. Basic financial features
    # =====================================================
    df['DailyProfit'] = df[sales_amount_col] - df[cost_amount_col]
    df['PriceCostRatio'] = df[unit_price_col] / (df[unit_cost_col] + 1e-5)
    df['ProfitMargin'] = df['DailyProfit'] / (df[sales_amount_col] + 1e-5)

    # =====================================================
    # 2. Calendar features
    # =====================================================
    df['Weekday'] = df[date_col].dt.dayofweek
    df['Day'] = df[date_col].dt.day
    df['Month'] = df[date_col].dt.month
    df['IsWeekend'] = df['Weekday'].isin([5, 6]).astype(int)

    doy = df[date_col].dt.dayofyear
    df['sin_doy'] = np.sin(2 * np.pi * doy / 365.25).astype('float32')
    df['cos_doy'] = np.cos(2 * np.pi * doy / 365.25).astype('float32')

    # =====================================================
    # 3. Static CV / Static sparsity
    # =====================================================
    if use_static:
        df_train_static = df[df[date_col].isin(train_days_list)].copy()

        min_date = df_train_static[date_col].min()
        max_date = df_train_static[date_col].max()
        n_days = (max_date - min_date).days + 1

        static_df = (
            df_train_static
            .groupby(item_col, observed=True)
            .agg(
                sum_qty=(target_col, 'sum'),
                sum_sq_qty=(target_col, lambda x: np.sum(x ** 2)),
                active_days=(target_col, lambda x: (x > 0).sum())
            )
            .reset_index()
        )

        static_df['avg_qty_real'] = static_df['sum_qty'] / max(n_days, 1)

        variance_real = np.maximum(
            static_df['sum_sq_qty'] / max(n_days, 1)
            - static_df['avg_qty_real'] ** 2,
            0
        )

        static_df['static_cv'] = (
            np.sqrt(variance_real) /
            (static_df['avg_qty_real'] + 1e-6)
        ).astype('float32')

        static_df['static_sparsity'] = (
            1 - static_df['active_days'] / max(n_days, 1)
        ).astype('float32')

        static_df = static_df[
            [item_col, 'static_cv', 'static_sparsity']
        ]

        global_static_cv = static_df['static_cv'].mean()
        global_static_sparsity = static_df['static_sparsity'].mean()

        df = df.merge(static_df, on=item_col, how='left')

        df['static_cv'] = df['static_cv'].fillna(global_static_cv)
        df['static_sparsity'] = df['static_sparsity'].fillna(global_static_sparsity)

    # =====================================================
    # 4. Lag features
    # =====================================================
    g = df.groupby(item_col, observed=True)

    lag_cols = []

    for lag in lags:
        col = f'lag_{lag}'
        log_col = f'lag_{lag}_log'

        df[col] = g[target_col].shift(lag)
        df[log_col] = np.log1p(df[col].fillna(0))

        lag_cols.extend([col, log_col])

    # =====================================================
    # 5. Rolling features
    # =====================================================
    rolling_cols = []

    for window in rolling_windows:
        mean_col = f'rolling_mean_{window}_lag1'
        std_col = f'rolling_std_{window}_lag1'
        max_col = f'rolling_max_{window}_lag1'
        min_col = f'rolling_min_{window}_lag1'

        df[mean_col] = (
            g[target_col]
            .transform(lambda x: x.shift(1).rolling(window, min_periods=1).mean())
        )

        df[std_col] = (
            g[target_col]
            .transform(lambda x: x.shift(1).rolling(window, min_periods=1).std())
        )

        df[max_col] = (
            g[target_col]
            .transform(lambda x: x.shift(1).rolling(window, min_periods=1).max())
        )

        df[min_col] = (
            g[target_col]
            .transform(lambda x: x.shift(1).rolling(window, min_periods=1).min())
        )

        rolling_cols.extend([mean_col, std_col, max_col, min_col])

    df['recent_trend'] = (
        df['rolling_mean_7_lag1'] /
        (df['rolling_mean_14_lag1'] + 1e-6)
    ).astype('float32')

    df['recent_trend'] = (
        df['recent_trend']
        .replace([np.inf, -np.inf], np.nan)
        .fillna(1.0)
    )

    # =====================================================
    # 6. Dynamic CV / sparsity
    # =====================================================
    df['SalesQty_tmp'] = df[target_col].clip(lower=0)
    df['QtySq_tmp'] = df['SalesQty_tmp'] ** 2
    df['IsZeroSale_tmp'] = (df['SalesQty_tmp'] == 0).astype(int)

    g = df.groupby(item_col, observed=True)

    cum_sales = g['SalesQty_tmp'].cumsum().groupby(df[item_col], observed=True).shift(1)
    cum_sq = g['QtySq_tmp'].cumsum().groupby(df[item_col], observed=True).shift(1)
    cum_zero = g['IsZeroSale_tmp'].cumsum().groupby(df[item_col], observed=True).shift(1)
    cum_days = (g.cumcount() + 1).groupby(df[item_col], observed=True).shift(1)

    mean_qty = cum_sales / cum_days
    variance_qty = (cum_sq / cum_days) - (mean_qty ** 2)
    variance_qty = np.maximum(variance_qty, 0)

    df['dynamic_cv_lag1'] = (
        np.sqrt(variance_qty) / (mean_qty + 1e-6)
    )

    df['dynamic_sparsity_lag1'] = cum_zero / cum_days

    df['dynamic_cv_lag1'] = (
        df['dynamic_cv_lag1']
        .replace([np.inf, -np.inf], np.nan)
        .fillna(0)
        .astype('float32')
    )

    df['dynamic_sparsity_lag1'] = (
        df['dynamic_sparsity_lag1']
        .replace([np.inf, -np.inf], np.nan)
        .fillna(1)
        .astype('float32')
    )

    df = df.drop(
        columns=['SalesQty_tmp', 'QtySq_tmp', 'IsZeroSale_tmp'],
        errors='ignore'
    )

    # =====================================================
    # 7. Days to last sell
    # =====================================================
    df['last_sell_date_tmp'] = np.where(
        df[target_col] > 0,
        df[date_col],
        pd.NaT
    )

    df['last_sell_date_tmp'] = pd.to_datetime(df['last_sell_date_tmp'])

    df['last_sell_date_tmp'] = (
        df.groupby(item_col, observed=True)['last_sell_date_tmp']
        .ffill()
    )

    df['days_to_last_sell'] = (
        df[date_col] - df['last_sell_date_tmp']
    ).dt.days

    df['days_to_last_sell'] = df['days_to_last_sell'].fillna(999)
    df = df.drop(columns=['last_sell_date_tmp'], errors='ignore')

    # =====================================================
    # 8. ISNA flags + fill
    # =====================================================
    dynamic_cols = []

    for lag in lags:
        dynamic_cols.append(f'lag_{lag}')

    dynamic_cols += rolling_cols

    for col in dynamic_cols:
        df[f'{col}_isna'] = df[col].isna().astype(int)
        df[col] = df[col].fillna(0)

    isna_cols = [f'{col}_isna' for col in dynamic_cols]

    for lag in lags:
        df[f'lag_{lag}_log'] = np.log1p(df[f'lag_{lag}'])

    # =====================================================
    # 9. Weight WRMSSE
    # =====================================================
    sku_profit = (
        df.groupby(item_col, observed=True)['DailyProfit']
        .sum()
        .reset_index()
    )

    sku_profit['clean_profit'] = np.where(
        sku_profit['DailyProfit'] < 0,
        0,
        sku_profit['DailyProfit']
    )

    total_system_profit = sku_profit['clean_profit'].sum()

    sku_profit['weight_wrmsse'] = (
        sku_profit['clean_profit'] / (total_system_profit + 1e-5)
    )

    df = df.drop(columns=['weight_wrmsse'], errors='ignore')

    df = df.merge(
        sku_profit[[item_col, 'weight_wrmsse']],
        on=item_col,
        how='left'
    )

    # =====================================================
    # 10. Feature list
    # =====================================================
    base_features = [
        item_col,
        'Weekday',
        'Day',
        'Month',
        'IsWeekend',
        'sin_doy',
        'cos_doy',
        'PriceCostRatio',
        'ProfitMargin',
        'days_to_last_sell',
        'dynamic_cv_lag1',
        'dynamic_sparsity_lag1',
        'recent_trend'
    ]

    if use_static:
        base_features += [
            'static_cv',
            'static_sparsity'
        ]

    lag_feature_cols = []

    for lag in lags:
        lag_feature_cols.append(f'lag_{lag}')
        lag_feature_cols.append(f'lag_{lag}_log')

    features = (
        base_features
        + lag_feature_cols
        + rolling_cols
        + isna_cols
    )

    df[item_col] = df[item_col].astype('category')

    output = {
        'df_daily': df,
        'features': features,
        'train_days_list': train_days_list,
        'val_days_list': val_days_list,
        'actual_days_sorted': actual_days_sorted
    }

    return output
```

## Model Class

```python

import lightgbm as lgb
import pandas as pd
import numpy as np


class LightGBMRecursiveModel:
    def __init__(
        self,
        features,
        target_col='TotalQuantity',
        item_col='ItemCode',
        date_col='Date',
        lags=(1, 2, 7, 14, 28),
        rolling_windows=(7, 14, 28),
        use_static=True,
        lgb_params=None
    ):
        self.features = features
        self.target_col = target_col
        self.item_col = item_col
        self.date_col = date_col
        self.lags = list(lags)
        self.rolling_windows = list(rolling_windows)
        self.use_static = use_static

        self.lgb_params = lgb_params if lgb_params is not None else {
            'objective': 'tweedie',
            'tweedie_variance_power': 1.15,
            'learning_rate': 0.05,
            'n_estimators': 700,
            'num_leaves': 64,
            'subsample': 0.8,
            'colsample_bytree': 0.8,
            'verbosity': -1,
            'n_jobs': -1,
            'random_state': 42
        }

        self.model = lgb.LGBMRegressor(**self.lgb_params)

    def _calculate_wrmsse(self, df_val_actual, preds, df_train_history, item_weights):
        df_train_history = df_train_history.sort_values(
            by=[self.item_col, self.date_col]
        ).copy()

        df_train_history['diff_sq'] = (
            df_train_history
            .groupby(self.item_col)[self.target_col]
            .diff()
            .pow(2)
        )

        scale_denom = (
            df_train_history
            .groupby(self.item_col)['diff_sq']
            .mean()
            .reset_index()
            .rename(columns={'diff_sq': 'denominator'})
        )

        scale_denom['denominator'] = (
            scale_denom['denominator']
            .replace(0, 1e-5)
            .fillna(1e-5)
        )

        df_eval = df_val_actual[[self.item_col, self.date_col, self.target_col]].copy()
        df_eval['pred'] = preds
        df_eval['se'] = (df_eval[self.target_col] - df_eval['pred']) ** 2

        val_mse = (
            df_eval
            .groupby(self.item_col)['se']
            .mean()
            .reset_index()
            .rename(columns={'se': 'mse'})
        )

        metric_df = val_mse.merge(scale_denom, on=self.item_col, how='left')
        metric_df['denominator'] = metric_df['denominator'].fillna(1e-5)

        metric_df['rmsse'] = np.sqrt(
            metric_df['mse'] / metric_df['denominator']
        )

        metric_df = metric_df.merge(item_weights, on=self.item_col, how='left')
        metric_df['weight_wrmsse'] = metric_df['weight_wrmsse'].fillna(0)

        return np.sum(metric_df['rmsse'] * metric_df['weight_wrmsse'])

    def _make_one_step_features(self, history_df, current_date):
        rows = []

        current_date = pd.to_datetime(current_date)

        for item, g in history_df.groupby(self.item_col, observed=False):
            g = g.sort_values(self.date_col)

            row = {
                self.item_col: item,
                self.date_col: current_date,
                'Weekday': current_date.dayofweek,
                'Day': current_date.day,
                'Month': current_date.month,
                'IsWeekend': int(current_date.dayofweek in [5, 6])
            }

            doy = current_date.dayofyear
            row['sin_doy'] = np.sin(2 * np.pi * doy / 365.25)
            row['cos_doy'] = np.cos(2 * np.pi * doy / 365.25)

            values = g[self.target_col].values

            for lag in self.lags:
                col = f'lag_{lag}'
                log_col = f'lag_{lag}_log'

                if len(values) >= lag:
                    row[col] = values[-lag]
                    row[log_col] = np.log1p(row[col])
                    row[f'{col}_isna'] = 0
                else:
                    row[col] = 0
                    row[log_col] = 0
                    row[f'{col}_isna'] = 1

            for window in self.rolling_windows:
                recent = values[-window:] if len(values) > 0 else []

                mean_col = f'rolling_mean_{window}_lag1'
                std_col = f'rolling_std_{window}_lag1'
                max_col = f'rolling_max_{window}_lag1'
                min_col = f'rolling_min_{window}_lag1'

                if len(recent) > 0:
                    row[mean_col] = np.mean(recent)
                    row[std_col] = np.std(recent, ddof=1) if len(recent) > 1 else 0
                    row[max_col] = np.max(recent)
                    row[min_col] = np.min(recent)
                else:
                    row[mean_col] = 0
                    row[std_col] = 0
                    row[max_col] = 0
                    row[min_col] = 0

                row[f'{mean_col}_isna'] = int(len(recent) == 0)
                row[f'{std_col}_isna'] = int(len(recent) <= 1)
                row[f'{max_col}_isna'] = int(len(recent) == 0)
                row[f'{min_col}_isna'] = int(len(recent) == 0)

            row['recent_trend'] = (
                row.get('rolling_mean_7_lag1', 0) /
                (row.get('rolling_mean_14_lag1', 0) + 1e-6)
            )

            row['recent_trend'] = (
                1.0 if not np.isfinite(row['recent_trend'])
                else row['recent_trend']
            )
            positive_sales = g[g[self.target_col] > 0]

            if len(positive_sales) > 0:
                last_sell_date = positive_sales[self.date_col].max()
                row['days_to_last_sell'] = (current_date - last_sell_date).days
            else:
                row['days_to_last_sell'] = 999

            sales = np.clip(values, 0, None)

            if len(sales) > 1:
                row['dynamic_sparsity_lag1'] = np.mean(sales == 0)

                mean_qty = np.mean(sales)
                std_qty = np.std(sales)

                row['dynamic_cv_lag1'] = (
                    std_qty / (mean_qty + 1e-6)
                )
            else:
                row['dynamic_sparsity_lag1'] = 1
                row['dynamic_cv_lag1'] = 0

            for col in ['PriceCostRatio', 'ProfitMargin']:
                if col in g.columns:
                    last_val = g[col].dropna()
                    row[col] = last_val.iloc[-1] if len(last_val) > 0 else 0
                else:
                    row[col] = 0

            if self.use_static:
                for col in ['static_cv', 'static_sparsity']:
                    if col in g.columns:
                        last_val = g[col].dropna()
                        row[col] = last_val.iloc[-1] if len(last_val) > 0 else 0
                    else:
                        row[col] = 0

            rows.append(row)

        df_features = pd.DataFrame(rows)

        for col in self.features:
            if col not in df_features.columns:
                df_features[col] = 0

        df_features[self.item_col] = df_features[self.item_col].astype('category')

        return df_features

    def predict_recursive(self, history_df, future_dates):
        history = history_df.copy()
        history[self.date_col] = pd.to_datetime(history[self.date_col])

        future_dates = sorted(pd.to_datetime(future_dates))

        all_preds = []

        for current_date in future_dates:
            X_step = self._make_one_step_features(history, current_date)

            preds = self.model.predict(X_step[self.features])
            preds = np.clip(preds, 0, None)

            pred_df = X_step[[self.item_col, self.date_col]].copy()
            pred_df[self.target_col] = preds

            for col in [
                'PriceCostRatio',
                'ProfitMargin',
                'days_to_last_sell',
                'dynamic_cv_lag1',
                'dynamic_sparsity_lag1',
                'static_cv',
                'static_sparsity'
            ]:
                if col in X_step.columns:
                    pred_df[col] = X_step[col].values

            all_preds.append(pred_df)

            history = pd.concat([history, pred_df], axis=0).reset_index(drop=True)
            history = history.sort_values([self.item_col, self.date_col])

        return pd.concat(all_preds, axis=0).reset_index(drop=True)

    def fit_giai_doan_1(self, df, train_days, val_days):
        print("====== GIAI ĐOẠN 1: TRAIN RECURSIVE LGBM + VALIDATE WRMSSE ======")

        df = df.copy()
        df[self.date_col] = pd.to_datetime(df[self.date_col])

        train_days = pd.to_datetime(train_days)
        val_days = pd.to_datetime(val_days)

        mask_train = df[self.date_col].isin(train_days)
        mask_val = df[self.date_col].isin(val_days)

        train_df = df.loc[mask_train].copy()
        val_df = df.loc[mask_val].copy()

        X_train = train_df[self.features]
        y_train = train_df[self.target_col].clip(lower=0)

        sample_weight = None
        if 'weight_wrmsse' in train_df.columns:
            sample_weight = train_df['weight_wrmsse'].values

        print("[Train] Đang huấn luyện LGBM recursive baseline...")

        self.model.fit(
            X_train,
            y_train,
            sample_weight=sample_weight,
            categorical_feature=[self.item_col]
        )

        print("[Validation] Đang forecast recursive trên tập validation...")

        pred_val_df = self.predict_recursive(
            history_df=train_df,
            future_dates=val_days
        )

        val_compare = val_df[[self.item_col, self.date_col, self.target_col]].merge(
            pred_val_df[[self.item_col, self.date_col, self.target_col]],
            on=[self.item_col, self.date_col],
            how='left',
            suffixes=('_actual', '_pred')
        )

        val_preds = val_compare[f'{self.target_col}_pred'].fillna(0).values

        item_weights = df[[self.item_col, 'weight_wrmsse']].drop_duplicates()

        wrmsse = self._calculate_wrmsse(
            df_val_actual=val_df,
            preds=val_preds,
            df_train_history=train_df,
            item_weights=item_weights
        )

        print(f"[Validation] WRMSSE recursive: {wrmsse:.6f}")

        return {
            'wrmsse_recursive': wrmsse,
            'pred_val_df': pred_val_df,
            'val_compare': val_compare
        }

    def retrain_giai_doan_2(self, df, history_days):
        print("\n====== GIAI ĐOẠN 2: RETRAIN FULL HISTORY ======")

        df = df.copy()
        df[self.date_col] = pd.to_datetime(df[self.date_col])
        history_days = pd.to_datetime(history_days)

        mask_history = df[self.date_col].isin(history_days)
        train_df = df.loc[mask_history].copy()

        X_full = train_df[self.features]
        y_full = train_df[self.target_col].clip(lower=0)

        sample_weight = None
        if 'weight_wrmsse' in train_df.columns:
            sample_weight = train_df['weight_wrmsse'].values

        self.model.fit(
            X_full,
            y_full,
            sample_weight=sample_weight,
            categorical_feature=[self.item_col]
        )

        print("Retrain recursive LGBM hoàn tất!")
```

## Full-grid training

```python
import pandas as pd
import numpy as np

def make_recent_full_grid_daily(
    df,
    days=180,
    date_col='Date',
    item_col='ItemCode',
    target_col='TotalQuantity',
    fill_zero_cols=None,
    ffill_cols=None
):
    df = df.copy()
    df[date_col] = pd.to_datetime(df[date_col])
    df[item_col] = df[item_col].astype(str)

    max_date = df[date_col].max()
    min_date = max_date - pd.Timedelta(days=days - 1)

    df_recent = df[df[date_col].between(min_date, max_date)].copy()

    all_items = df_recent[item_col].unique()
    all_dates = pd.date_range(min_date, max_date, freq='D')

    full_index = pd.MultiIndex.from_product(
        [all_items, all_dates],
        names=[item_col, date_col]
    )

    df_full = (
        df_recent
        .set_index([item_col, date_col])
        .reindex(full_index)
        .reset_index()
    )

    if fill_zero_cols is None:
        fill_zero_cols = [
            target_col,
            'TotalSalesAmount',
            'TotalCostAmount'
        ]

    for col in fill_zero_cols:
        if col in df_full.columns:
            df_full[col] = df_full[col].fillna(0)

    if ffill_cols is None:
        ffill_cols = [
            'AverageUnitPrice',
            'AverageUnitCost'
        ]

    for col in ffill_cols:
        if col in df_full.columns:
            df_full[col] = (
                df_full.groupby(item_col)[col]
                .ffill()
                .bfill()
                .fillna(0)
            )

    df_full = df_full.sort_values([item_col, date_col]).reset_index(drop=True)

    print(f"Recent full-grid range: {min_date.date()} -> {max_date.date()}")
    print(f"Items: {len(all_items):,}")
    print(f"Rows: {df_full.shape[0]:,}")

    return df_full
```

```python
df_grouped_full = make_recent_full_grid_daily(
    df=df_grouped,
    days=270,
    date_col='Date',
    item_col='ItemCode',
    target_col='TotalQuantity'
)

fe_output = build_recursive_train_features(
    df_grouped=df_grouped_full,
    lags=(1, 2, 7, 14, 28),
    rolling_windows=(7, 14, 28),
    val_size=28,
    use_static=True
)

df_fe = fe_output['df_daily']
features = fe_output['features']
train_days = fe_output['train_days_list']
val_days = fe_output['val_days_list']
actual_days_sorted = fe_output['actual_days_sorted']
```

```python
recursive_model_grid = LightGBMRecursiveModel(
    features=features,
    lags=(1, 2, 7, 14, 28),
    rolling_windows=(7, 14, 28),
    use_static=True,
    target_col='TotalQuantity'
)

scores = recursive_model_grid.fit_giai_doan_1(
    df=df_fe,
    train_days=train_days,
    val_days=val_days
)

print(scores['wrmsse_recursive'])
```

```python
import numpy as np
import pandas as pd


def make_submission_from_recursive_model(
    df_daily,
    recursive_model,
    sample_submission_path='sample_submission.csv',
    date_col='Date',
    item_col='ItemCode',
    target_col='TotalQuantity'
):
    df_template = pd.read_csv(sample_submission_path)

    id_col = 'id' if 'id' in df_template.columns else item_col

    df_template['clean_sku'] = (
        df_template[id_col]
        .astype(str)
        .str.replace('_validation', '', regex=False)
        .str.replace('_evaluation', '', regex=False)
    )

    unique_skus = df_template['clean_sku'].unique()

    history_df = df_daily.copy()
    history_df[date_col] = pd.to_datetime(history_df[date_col])
    history_df[item_col] = history_df[item_col].astype(str)

    history_df = history_df[
        history_df[item_col].isin(unique_skus)
    ].copy()

    max_train_date = history_df[date_col].max()

    future_dates = [
        max_train_date + pd.to_timedelta(i, unit='D')
        for i in range(1, 57)
    ]

    validation_dates = future_dates[:28]
    evaluation_dates = future_dates[28:]

    pred_56_df = recursive_model.predict_recursive(
        history_df=history_df,
        future_dates=future_dates
    )

    pred_56_df[target_col] = pred_56_df[target_col].clip(lower=0)

    df_public = pred_56_df[
        pred_56_df[date_col].isin(validation_dates)
    ].copy()

    df_private = pred_56_df[
        pred_56_df[date_col].isin(evaluation_dates)
    ].copy()

    def pivot_to_wide_format(df_partition, suffix):
        df_partition = df_partition.copy()
        df_partition[item_col] = df_partition[item_col].astype(str)

        df_partition = df_partition.sort_values([item_col, date_col])
        df_partition['day_idx'] = (
            df_partition.groupby(item_col).cumcount() + 1
        )

        df_partition['F_col'] = 'F' + df_partition['day_idx'].astype(str)

        f_cols = [f'F{i}' for i in range(1, 29)]

        df_pivot = (
            df_partition
            .pivot(index=item_col, columns='F_col', values=target_col)
            .reindex(columns=f_cols)
            .reset_index()
        )

        df_pivot['id'] = (
            df_pivot[item_col].astype(str) + suffix
        )

        return df_pivot[['id'] + f_cols]

    df_validation_wide = pivot_to_wide_format(
        df_public,
        suffix='_validation'
    )

    df_evaluation_wide = pivot_to_wide_format(
        df_private,
        suffix='_evaluation'
    )

    df_preds_wide = pd.concat(
        [df_validation_wide, df_evaluation_wide],
        axis=0
    ).reset_index(drop=True)

    f_cols = [f'F{i}' for i in range(1, 29)]

    df_final_submission = (
        df_template[['id']]
        .merge(df_preds_wide, on='id', how='left')
    )

    df_final_submission[f_cols] = (
        df_final_submission[f_cols]
        .fillna(0)
        .clip(lower=0)
    )

    assert df_final_submission.shape[0] == df_template.shape[0]
    assert df_final_submission[f_cols].isna().sum().sum() == 0

    print("Tạo submission recursive hoàn tất!")

    return df_final_submission, pred_56_df
```
