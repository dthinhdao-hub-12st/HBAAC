```python
def check_date_range(df, stage_name, expected_days=None):
    # Đảm bảo cột Date đã ở định dạng datetime
    dates = pd.to_datetime(df['Date'])

    # Lấy ngày nhỏ nhất, lớn nhất và đếm số lượng ngày độc bản
    min_date = dates.min().strftime('%Y-%m-%d')
    max_date = dates.max().strftime('%Y-%m-%d')
    unique_days = dates.nunique()

    print(f"GIAI ĐOẠN: {stage_name.upper()}")
    print(f"Từ ngày:       {min_date}")
    print(f"Đến ngày:      {max_date}")
    print(f"Số ngày thực tế: {unique_days}")

    # Nếu có số ngày kỳ vọng, tiến hành kiểm tra tự động
    if expected_days is not None:
        if unique_days == expected_days:
            print("HỢP LỆ: Đủ số lượng ngày cấu hình.")
        else:
            print(f"LỖI: Thiếu hoặc thừa dữ liệu! Kỳ vọng {expected_days} ngày nhưng thực tế có {unique_days} ngày.")
    print("-" * 40)

# 1. Kiểm tra tập Train (Từ ngày đầu tiên đến hết 08/08/2025)
check_date_range(train_data_copy[train_data_copy['Date'] <= '2025-08-08'], "Train")

# 2. Kiểm tra tập Valid (Phải đủ 28 ngày: 09/08/2025 -> 05/09/2025)
valid_df = train_data_copy[(train_data_copy['Date'] >= '2025-08-09') & (train_data_copy['Date'] <= '2025-09-05')]
check_date_range(valid_df, "Validation", expected_days=28)
```

# Model Baseline (LGBM)

## Feature Engineering

```python
import numpy as np
import pandas as pd

def prepare_daily_features_and_split_baseline(
    df_grouped,
    date_col='Date',
    item_col='ItemCode',
    target_col='TotalQuantity',
    val_size=28
):
    df_daily = df_grouped.sort_values(by=[item_col, date_col]).reset_index(drop=True).copy()

    df_daily[item_col] = df_daily[item_col].astype('category')
    df_daily[date_col] = pd.to_datetime(df_daily[date_col])

    # =========================
    # 1. Profit features
    # =========================
    df_daily['DailyProfit'] = df_daily['TotalSalesAmount'] - df_daily['TotalCostAmount']
    df_daily['PriceCostRatio'] = df_daily['AverageUnitPrice'] / (df_daily['AverageUnitCost'] + 1e-5)
    df_daily['ProfitMargin'] = df_daily['DailyProfit'] / (df_daily['TotalSalesAmount'] + 1e-5)

    # =========================
    # 2. Time features
    # =========================
    df_daily['Weekday'] = df_daily[date_col].dt.dayofweek
    df_daily['Day'] = df_daily[date_col].dt.day
    df_daily['Month'] = df_daily[date_col].dt.month
    df_daily['IsWeekend'] = df_daily['Weekday'].isin([5, 6]).astype(int)

    # =========================
    # 3. Lag features
    # =========================
    groupby_item = df_daily.groupby(item_col, observed=False)

    for lag in [28, 29, 35, 56, 57, 63]:
        df_daily[f'lag_{lag}'] = groupby_item[target_col].shift(lag)

    # =========================
    # 4. Rolling features
    # =========================
    rolling_specs = [
        (28, 7, 'mean'),
        (28, 7, 'std'),
        (28, 30, 'mean'),
        (56, 7, 'mean'),
        (56, 7, 'std'),
        (56, 14, 'std'),
        (56, 14, 'mean'),
        (56, 30, 'mean'),
    ]

    for lag, window, stat in rolling_specs:
        col_name = f'rolling_{stat}_{window}_lag{lag}'

        if stat == 'mean':
            df_daily[col_name] = (
                df_daily.groupby(item_col, observed=False)[target_col]
                .transform(lambda x: x.shift(lag).rolling(window, min_periods=1).mean())
            )
        elif stat == 'std':
            df_daily[col_name] = (
                df_daily.groupby(item_col, observed=False)[target_col]
                .transform(lambda x: x.shift(lag).rolling(window, min_periods=1).std())
            )

    # =========================
    # 5. NaN indicator + fill
    # =========================
    lag_cols = [
        'lag_28', 'lag_29', 'lag_35',
        'lag_56', 'lag_57', 'lag_63',
        'rolling_mean_7_lag28',
        'rolling_std_7_lag28',
        'rolling_mean_30_lag28',
        'rolling_mean_7_lag56',
        'rolling_std_7_lag56',
        'rolling_std_14_lag56',
        'rolling_mean_14_lag56',
        'rolling_mean_30_lag56',
    ]

    for col in lag_cols:
        df_daily[f'{col}_isna'] = df_daily[col].isna().astype(int)
        df_daily[col] = df_daily[col].fillna(0)

    # =========================
    # 6. Days to last sell
    # =========================
    df_daily['last_sell_date'] = np.where(
        df_daily[target_col] > 0,
        df_daily[date_col],
        pd.NaT
    )

    df_daily['last_sell_date'] = pd.to_datetime(df_daily['last_sell_date'])
    df_daily['last_sell_date'] = (
        df_daily.groupby(item_col, observed=False)['last_sell_date']
        .ffill()
    )

    df_daily['days_to_last_sell'] = (
        df_daily[date_col] - df_daily['last_sell_date']
    ).dt.days

    df_daily['days_to_last_sell'] = df_daily['days_to_last_sell'].fillna(999)
    df_daily = df_daily.drop(columns=['last_sell_date'])

    # =========================
    # 7. Feature lists
    # =========================
    base_features = [
        item_col, 'Weekday', 'Day', 'Month', 'IsWeekend',
        'days_to_last_sell', 'PriceCostRatio', 'ProfitMargin'
    ]

    features_for_public = base_features + [
        'lag_28', 'lag_29', 'lag_35',
        'rolling_mean_7_lag28',
        'rolling_std_7_lag28',
    ] + [
        'lag_28_isna', 'lag_29_isna', 'lag_35_isna',
        'rolling_mean_7_lag28_isna',
        'rolling_std_7_lag28_isna',
    ]

    features_for_private = base_features + [
        'lag_56', 'lag_57',
        'rolling_mean_7_lag56',
        'rolling_std_7_lag56',
    ] + [
        'lag_56_isna', 'lag_57_isna',
        'rolling_mean_7_lag56_isna',
        'rolling_std_7_lag56_isna',
    ]

    # =========================
    # 8. Train / validation split
    # =========================
    actual_days_sorted = sorted(df_daily[date_col].unique())

    val_days_list = actual_days_sorted[-val_size:]
    train_days_list = actual_days_sorted[:-val_size]

    train_mask = df_daily[date_col].isin(train_days_list)
    valid_mask = df_daily[date_col].isin(val_days_list)

    X_train_public = df_daily.loc[train_mask, features_for_public]
    y_train_public = df_daily.loc[train_mask, target_col]

    X_val_public = df_daily.loc[valid_mask, features_for_public]
    y_val_public = df_daily.loc[valid_mask, target_col]

    X_train_private = df_daily.loc[train_mask, features_for_private]
    y_train_private = df_daily.loc[train_mask, target_col]

    X_val_private = df_daily.loc[valid_mask, features_for_private]
    y_val_private = df_daily.loc[valid_mask, target_col]

    # =========================
    # 9. WRMSSE-like weight
    # =========================
    sku_profit = (
        df_daily.groupby(item_col, observed=False)['DailyProfit']
        .sum()
        .reset_index()
    )

    sku_profit['clean_profit'] = np.where(
        sku_profit['DailyProfit'] < 0,
        0,
        sku_profit['DailyProfit']
    )

    total_system_profit = sku_profit['clean_profit'].sum()

    sku_profit['weight_wrmsse'] = (
        sku_profit['clean_profit'] / (total_system_profit + 1e-5)
    )

    df_daily = df_daily.drop(columns=['weight_wrmsse'], errors='ignore')
    df_daily = df_daily.merge(
        sku_profit[[item_col, 'weight_wrmsse']],
        on=item_col,
        how='left'
    )

    # =========================================================
    # 10. SKU GROUP
    # =========================================================
    sku_group = (
        df_daily[df_daily[date_col].isin(train_days_list)]
        .groupby(item_col, observed=True)
        .agg(
            total_profit=('DailyProfit', 'sum'),
            total_qty=(target_col, 'sum'),
            active_days=(target_col, lambda x: (x > 0).sum()),
            n_days=(date_col, 'nunique')
        )
        .reset_index()
    )

    sku_group['positive_profit'] = sku_group['total_profit'].clip(lower=0)

    last_sell = (
        df_daily[
            (df_daily[date_col].isin(train_days_list)) &
            (df_daily[target_col] > 0)
        ]
        .groupby(item_col, observed=True)[date_col]
        .max()
        .reset_index()
        .rename(columns={date_col: 'last_sell_date'})
    )

    sku_group = sku_group.merge(last_sell, on=item_col, how='left')
    max_train_date = max(train_days_list)
    sku_group['days_since_last_sell'] = (max_train_date - sku_group['last_sell_date']).dt.days
    sku_group['days_since_last_sell'] = sku_group['days_since_last_sell'].fillna(9999)
    sku_group['active_ratio'] = sku_group['active_days'] / sku_group['n_days'].clip(lower=1)

    def assign_sku_group(row):
        pp = row['positive_profit']
        if pp >= 28.28e6: return 'A'
        if pp >= 6.94e6: return 'B'
        if pp >= 2.58e6: return 'C'
        if (pp > 0) and (row['days_since_last_sell'] >= 90): return 'Dead'
        if (pp > 0) and (row['active_ratio'] <= 0.10): return 'Slow'
        if pp > 0: return 'Medium'
        return 'Profit <= 0'

    sku_group['sku_group'] = sku_group.apply(assign_sku_group, axis=1)

    df_daily = df_daily.merge(
        sku_group[[item_col, 'positive_profit', 'sku_group']],
        on=item_col,
        how='left'
    )

    df_daily['positive_profit'] = df_daily['positive_profit'].fillna(0)
    df_daily['sku_group'] = df_daily['sku_group'].fillna('Profit <= 0')
    df_daily['sku_group'] = df_daily['sku_group'].astype('category')

    print("\nChia dữ liệu hoàn tất!")

    return {
        'df_daily': df_daily,
        'features_for_public': features_for_public,
        'features_for_private': features_for_private,
        'X_train_public': X_train_public,
        'y_train_public': y_train_public,
        'X_val_public': X_val_public,
        'y_val_public': y_val_public,
        'X_train_private': X_train_private,
        'y_train_private': y_train_private,
        'X_val_private': X_val_private,
        'y_val_private': y_val_private,
        'train_days_list': train_days_list,
        'val_days_list': val_days_list,
        'actual_days_sorted': actual_days_sorted,
    }
```

## Model Class (Cơ bản)

```python
import lightgbm as lgb
import pandas as pd
import numpy as np

class LightGBMTwinModel:
    def __init__(self, features_public, features_private, target_col='TotalQuantity', lgb_params=None):
        self.features_public = features_public
        self.features_private = features_private
        self.target_col = target_col

        self.lgb_params = lgb_params if lgb_params is not None else {
            'objective': 'tweedie',
            'tweedie_variance_power': 1.15,
            'learning_rate': 0.05,
            'n_estimators': 500,
            'verbosity': -1,
            'n_jobs': -1,
            'random_state': 42
        }

        self.model_public = lgb.LGBMRegressor(**self.lgb_params)
        self.model_private = lgb.LGBMRegressor(**self.lgb_params)

        # Add flags to indicate if a model should always predict zero
        self.is_zero_predictor_public = False
        self.is_zero_predictor_private = False

    def _calculate_wrmsse(self, df_val, preds, df_train_history, item_weights):
      df_train_history = df_train_history.sort_values(by=['ItemCode', 'Date']).copy()

      df_train_history['diff_sq'] = (
          df_train_history
          .groupby('ItemCode', observed=False)[self.target_col]
          .diff()
          .pow(2)
      )

      scale_denom = (
          df_train_history
          .groupby('ItemCode', observed=False)['diff_sq']
          .mean()
          .reset_index()
          .rename(columns={'diff_sq': 'denominator'})
      )

      scale_denom['denominator'] = (
          scale_denom['denominator']
          .replace(0, 1e-5)
          .fillna(1e-5)
      )

      df_eval = df_val[['ItemCode', self.target_col]].copy()
      df_eval['pred'] = preds
      df_eval['se'] = (df_eval[self.target_col] - df_eval['pred']) ** 2

      val_mse = (
          df_eval
          .groupby('ItemCode', observed=False)['se']
          .mean()
          .reset_index()
          .rename(columns={'se': 'mse'})
      )

      metric_df = val_mse.merge(scale_denom, on='ItemCode', how='left')
      metric_df['denominator'] = metric_df['denominator'].fillna(1e-5)

      metric_df['rmsse'] = np.sqrt(metric_df['mse'] / metric_df['denominator'])

      metric_df = metric_df.merge(item_weights, on='ItemCode', how='left')
      metric_df['weight_wrmsse'] = metric_df['weight_wrmsse'].fillna(0)

      wrmsse_score = np.sum(metric_df['rmsse'] * metric_df['weight_wrmsse'])

      return wrmsse_score

    def fit_giai_doan_1(self, df, train_days, val_days, date_col='Date'):
        print("====== BẮT ĐẦU GIAI ĐOẠN 1: HUẤN LUYỆN & TÍNH ĐIỂM WRMSSE ======")
        mask_train_base = df[date_col].isin(train_days)
        mask_val = df[date_col].isin(val_days)

        # Trích xuất bảng trọng số độc bản của từng SKU để phục vụ tính điểm
        item_weights = df[['ItemCode', 'weight_wrmsse']].drop_duplicates()

        # --- 1. Huấn luyện và Đánh giá nhánh Public ---
        print("[Public] Đang huấn luyện với trọng số tích lũy...")
        mask_train_pub = mask_train_base & df['rolling_mean_7_lag28'].notna()

        X_train_pub = df[mask_train_pub][self.features_public]
        y_train_pub = df[mask_train_pub][self.target_col].clip(lower=0) # Clip target to be non-negative
        w_train_pub = df[mask_train_pub]['weight_wrmsse'].values

        if y_train_pub.sum() == 0:
            print(f"[Public] Target variable for this group is all zeros. Skipping training and setting to predict 0.")
            self.is_zero_predictor_public = True
            wrmsse_pub = 0.0 # No sales, so WRMSSE is 0
        else:
            self.model_public.fit(X_train_pub, y_train_pub, sample_weight=w_train_pub)
            self.is_zero_predictor_public = False

            # Dự đoán tập Val và tính điểm chuẩn metric cuộc thi
            X_val_pub = df[mask_val][self.features_public]
            pred_val_pub = self.model_public.predict(X_val_pub).clip(0, None)

            wrmsse_pub = self._calculate_wrmsse(df[mask_val], pred_val_pub, df[mask_train_pub], item_weights)
        print(f"[Public] Hoàn tất! ĐIỂM WRMSSE trên tập Validation: {wrmsse_pub:.6f}")

        # --- 2. Huấn luyện và Đánh giá nhánh Private ---
        print("\n[Private] Đang huấn luyện với trọng số tích lũy...")
        mask_train_pri = mask_train_base & df['rolling_mean_7_lag56'].notna()

        X_train_pri = df[mask_train_pri][self.features_private]
        y_train_pri = df[mask_train_pri][self.target_col].clip(lower=0) # Clip target to be non-negative
        w_train_pri = df[mask_train_pri]['weight_wrmsse'].values

        if y_train_pri.sum() == 0:
            print(f"[Private] Target variable for this group is all zeros. Skipping training and setting to predict 0.")
            self.is_zero_predictor_private = True
            wrmsse_pri = 0.0 # No sales, so WRMSSE is 0
        else:
            self.model_private.fit(X_train_pri, y_train_pri, sample_weight=w_train_pri)
            self.is_zero_predictor_private = False

            # Dự đoán tập Val và tính điểm chuẩn cho Private
            X_val_pri = df[mask_val][self.features_private]
            pred_val_pri = self.model_private.predict(X_val_pri).clip(0, None)

            wrmsse_pri = self._calculate_wrmsse(df[mask_val], pred_val_pri, df[mask_train_pri], item_weights)
        print(f"[Private] Hoàn tất! ĐIỂM WRMSSE trên tập Validation: {wrmsse_pri:.6f}")

        return {"wrmsse_public": wrmsse_pub, "wrmsse_private": wrmsse_pri}

    def retrain_giai_doan_2(self, df, history_days, date_col='Date'):
        print("\n====== BẮT ĐẦU GIAI ĐOẠN 2: TỔNG LỰC RETRAIN ======")
        mask_history_base = df[date_col].isin(history_days)

        # --- 1. Retrain Public ---
        print("[Public] Đang tái hấp thụ toàn bộ lịch sử...")
        mask_train_pub = mask_history_base & df['rolling_mean_7_lag28'].notna()
        X_full_pub = df[mask_train_pub][self.features_public]
        y_full_pub = df[mask_train_pub][self.target_col].clip(lower=0) # Clip target to be non-negative
        w_full_pub = df[mask_train_pub]['weight_wrmsse'].values

        if y_full_pub.sum() == 0:
            print(f"[Public] Target variable for this group is all zeros. Skipping retraining.")
            self.is_zero_predictor_public = True
        else:
            self.model_public.fit(X_full_pub, y_full_pub, sample_weight=w_full_pub)
            self.is_zero_predictor_public = False

        # --- 2. Retrain Private ---
        print("[Private] Đang tái hấp thụ toàn bộ lịch sử...")
        mask_train_pri = mask_history_base & df['rolling_mean_7_lag56'].notna()
        X_full_pri = df[mask_train_pri][self.features_private]
        y_full_pri = df[mask_train_pri][self.target_col].clip(lower=0) # Clip target to be non-negative
        w_full_pri = df[mask_train_pri]['weight_wrmsse'].values

        if y_full_pri.sum() == 0:
            print(f"[Private] Target variable for this group is all zeros. Skipping retraining.")
            self.is_zero_predictor_private = True
        else:
            self.model_private.fit(X_full_pri, y_full_pri, sample_weight=w_full_pri)
            self.is_zero_predictor_private = False

        print("Đã chuẩn bị xong bộ đôi LightGBM thực chiến!")

    def predict_submission(self, df_test_public, df_test_private):
        print("\nĐang tiến hành dự đoán tập Test thực tế...")

        if self.is_zero_predictor_public:
            preds_pub = np.zeros(len(df_test_public))
        else:
            preds_pub = self.model_public.predict(df_test_public[self.features_public])

        if self.is_zero_predictor_private:
            preds_pri = np.zeros(len(df_test_private))
        else:
            preds_pri = self.model_private.predict(df_test_private[self.features_private])

        preds_pub_clean = np.clip(preds_pub, a_min=0, a_max=None)
        preds_pri_clean = np.clip(preds_pri, a_min=0, a_max=None)

        return preds_pub_clean, preds_pri_clean
```

```python
def make_recent_full_grid_daily(
    df,
    days=180,
    date_col='Date',
    item_col='ItemCode',
    target_col='TotalQuantity',
    fill_zero_cols=None,
    ffill_cols=None
):
    df = df.copy()
    df[date_col] = pd.to_datetime(df[date_col])
    df[item_col] = df[item_col].astype(str)

    max_date = df[date_col].max()
    min_date = max_date - pd.Timedelta(days=days - 1)

    df_recent = df[df[date_col].between(min_date, max_date)].copy()

    all_items = df_recent[item_col].unique()
    all_dates = pd.date_range(min_date, max_date, freq='D')

    full_index = pd.MultiIndex.from_product(
        [all_items, all_dates],
        names=[item_col, date_col]
    )

    df_full = (
        df_recent
        .set_index([item_col, date_col])
        .reindex(full_index)
        .reset_index()
    )

    if fill_zero_cols is None:
        fill_zero_cols = [
            target_col,
            'TotalSalesAmount',
            'TotalCostAmount'
        ]

    for col in fill_zero_cols:
        if col in df_full.columns:
            df_full[col] = df_full[col].fillna(0)

    if ffill_cols is None:
        ffill_cols = [
            'AverageUnitPrice',
            'AverageUnitCost'
        ]

    for col in ffill_cols:
        if col in df_full.columns:
            df_full[col] = (
                df_full.groupby(item_col)[col]
                .ffill()
                .bfill()
                .fillna(0)
            )

    df_full = df_full.sort_values([item_col, date_col]).reset_index(drop=True)

    print(f"Recent full-grid range: {min_date.date()} -> {max_date.date()}")
    print(f"Items: {len(all_items):,}")
    print(f"Rows: {df_full.shape[0]:,}")

    return df_full
```

```python
# 1. Hàm tính toán và tổng hợp điểm số từ Dictionary các model
def calculate_multiple_model_scores(group_models, df_full, val_days):
    results = []
    total_wrmsse_public = 0
    total_wrmsse_private = 0

    mask_val = df_full['Date'].isin(val_days)

    print(f"{'Group':<15} | {'Public WRMSSE':<15} | {'Private WRMSSE':<15}")
    print("-" * 50)

    for group_name, model in group_models.items():
        # Lấy subset dữ liệu của nhóm này trong tập validation
        df_group_val = df_full[mask_val & (df_full['sku_group'] == group_name)]

        if df_group_val.empty:
            continue

        # Dự đoán cho nhánh Public
        preds_pub = model.model_public.predict(df_group_val[model.features_public]).clip(0, None)
        # Dự đoán cho nhánh Private
        preds_pri = model.model_private.predict(df_group_val[model.features_private]).clip(0, None)

        # Lưu ý: Hàm _calculate_wrmsse trong class trả về tổng (score * weight) của subset đó
        # nên chúng ta chỉ cần cộng dồn các giá trị này lại.
        score_pub = model._calculate_wrmsse(
            df_val=df_group_val,
            preds=preds_pub,
            df_train_history=df_full[df_full['Date'].isin(train_days_list) & (df_full['sku_group'] == group_name)],
            item_weights=df_full[['ItemCode', 'weight_wrmsse']].drop_duplicates()
        )

        score_pri = model._calculate_wrmsse(
            df_val=df_group_val,
            preds=preds_pri,
            df_train_history=df_full[df_full['Date'].isin(train_days_list) & (df_full['sku_group'] == group_name)],
            item_weights=df_full[['ItemCode', 'weight_wrmsse']].drop_duplicates()
        )

        total_wrmsse_public += score_pub
        total_wrmsse_private += score_pri

        print(f"{group_name:<15} | {score_pub:<15.6f} | {score_pri:<15.6f}")

    print("-" * 50)
    print(f"{'TOTAL SYSTEM':<15} | {total_wrmsse_public:<15.6f} | {total_wrmsse_private:<15.6f}")

    return total_wrmsse_public, total_wrmsse_private

# 2. Chạy tính toán điểm
final_pub_score, final_pri_score = calculate_multiple_model_scores(
    group_models=group_models,
    df_full=df_daily_baseline,
    val_days=val_days_list
)
```

```python
import numpy as np
import pandas as pd

def make_submission_from_multiple_models(
    df_daily,
    group_models,
    features_for_public,
    features_for_private,
    sample_submission_path='sample_submission.csv',
    date_col='Date',
    item_col='ItemCode',
    target_col='TotalQuantity'
):
    # =========================
    # 1. Đọc sample submission
    # =========================
    df_test_template = pd.read_csv(sample_submission_path)

    id_col = 'id' if 'id' in df_test_template.columns else item_col

    df_test_template['clean_sku'] = (
        df_test_template[id_col]
        .astype(str)
        .str.replace('_validation', '', regex=False)
        .str.replace('_evaluation', '', regex=False)
    )

    unique_skus = df_test_template['clean_sku'].unique()

    # =========================
    # 2. Tạo 56 ngày tương lai
    # =========================
    df_daily = df_daily.copy()
    df_daily[date_col] = pd.to_datetime(df_daily[date_col])

    max_train_date = pd.to_datetime(df_daily[date_col].max())

    future_dates = [
        max_train_date + pd.to_timedelta(i, unit='D')
        for i in range(1, 57)
    ]

    validation_dates = future_dates[:28]
    evaluation_dates = future_dates[28:]

    df_test_clean_vertical = (
        pd.MultiIndex
        .from_product([unique_skus, future_dates], names=[item_col, date_col])
        .to_frame(index=False)
    )

    df_test_clean_vertical[target_col] = np.nan
    df_test_clean_vertical['is_test'] = 1

    df_daily['is_test'] = 0

    # =========================
    # 3. Gộp lịch sử + test
    # =========================
    df_daily[item_col] = df_daily[item_col].astype(str)
    df_test_clean_vertical[item_col] = df_test_clean_vertical[item_col].astype(str)

    df_master = pd.concat(
        [df_daily, df_test_clean_vertical],
        axis=0
    ).reset_index(drop=True)

    df_master = (
        df_master
        .sort_values(by=[item_col, date_col])
        .reset_index(drop=True)
    )

    # =========================
    # 4. Time features
    # =========================
    df_master['Weekday'] = df_master[date_col].dt.weekday
    df_master['Day']     = df_master[date_col].dt.day
    df_master['Month']   = df_master[date_col].dt.month
    df_master['IsWeekend'] = (df_master['Weekday'] >= 5).astype(int)

    # =========================
    # 5. Lags public / private
    # =========================
    groupby_item = df_master.groupby(item_col, observed=False)

    df_master['lag_28'] = groupby_item[target_col].shift(28)
    df_master['lag_29'] = groupby_item[target_col].shift(29)
    df_master['lag_35'] = groupby_item[target_col].shift(35)
    df_master['lag_56'] = groupby_item[target_col].shift(56)
    df_master['lag_57'] = groupby_item[target_col].shift(57)

    # =========================
    # 6. Rolling features
    # =========================
    df_master['rolling_mean_7_lag28'] = (
        df_master.groupby(item_col, observed=False)[target_col]
        .transform(lambda x: x.shift(28).rolling(7, min_periods=1).mean())
    )
    df_master['rolling_std_7_lag28'] = (
        df_master.groupby(item_col, observed=False)[target_col]
        .transform(lambda x: x.shift(28).rolling(7, min_periods=1).std())
    )
    df_master['rolling_mean_7_lag56'] = (
        df_master.groupby(item_col, observed=False)[target_col]
        .transform(lambda x: x.shift(56).rolling(7, min_periods=1).mean())
    )
    df_master['rolling_std_7_lag56'] = (
        df_master.groupby(item_col, observed=False)[target_col]
        .transform(lambda x: x.shift(56).rolling(7, min_periods=1).std())
    )

    # =========================
    # 7. NaN indicator + fill
    # =========================
    lag_cols = [
        'lag_28', 'lag_29', 'lag_35',
        'lag_56', 'lag_57',
        'rolling_mean_7_lag28', 'rolling_std_7_lag28',
        'rolling_mean_7_lag56', 'rolling_std_7_lag56'
    ]
    for col in lag_cols:
        df_master[f'{col}_isna'] = df_master[col].isna().astype(int)
        df_master[col] = df_master[col].fillna(0)

    # =========================
    # 8. Fill financial static features
    # =========================
    df_master['PriceCostRatio'] = (
        df_master.groupby(item_col, observed=False)['PriceCostRatio'].ffill()
    )
    df_master['ProfitMargin'] = (
        df_master.groupby(item_col, observed=False)['ProfitMargin'].ffill()
    )

    # =========================
    # 9. Recalculate days_to_last_sell
    # =========================
    df_master['last_sell_date_calculated'] = np.where(
        df_master['is_test'] == 0,
        df_master[date_col] - pd.to_timedelta(df_master['days_to_last_sell'], unit='D'),
        pd.NaT
    )
    df_master['last_sell_date_calculated'] = pd.to_datetime(
        df_master['last_sell_date_calculated']
    )
    df_master['last_sell_date_calculated'] = (
        df_master.groupby(item_col, observed=False)['last_sell_date_calculated'].ffill()
    )
    df_master['days_to_last_sell'] = (
        df_master[date_col] - df_master['last_sell_date_calculated']
    ).dt.days
    df_master['days_to_last_sell'] = df_master['days_to_last_sell'].fillna(999)
    df_master = df_master.drop(columns=['last_sell_date_calculated'])

    sku_to_group = (
        df_daily[[item_col, 'sku_group']]
        .drop_duplicates(subset=[item_col])
        .set_index(item_col)['sku_group']
        .to_dict()
    )
    df_master['sku_group'] = df_master[item_col].map(sku_to_group)

    # SKU xuất hiện trong sample_submission nhưng không có trong lịch sử gán vào nhóm fallback (ví dụ 'Dead')
    FALLBACK_GROUP = 'Dead'
    n_unknown = df_master['sku_group'].isna().sum()
    if n_unknown > 0:
        print(f"[WARN] {n_unknown} dòng không có sku_group → gán fallback='{FALLBACK_GROUP}'")
    df_master['sku_group'] = df_master['sku_group'].fillna(FALLBACK_GROUP)

    # =========================
    # 10. Lọc ra tập test + ép category
    # =========================
    df_test_all = df_master[df_master['is_test'] == 1].copy()

    unique_test_days = sorted(df_test_all[date_col].unique())
    assert len(unique_test_days) == 56, \
        f"Số ngày test hiện tại = {len(unique_test_days)}, cần đúng 56"

    public_days  = unique_test_days[:28]
    private_days = unique_test_days[28:]

    df_test_public_all  = df_test_all[df_test_all[date_col].isin(public_days)].copy()
    df_test_private_all = df_test_all[df_test_all[date_col].isin(private_days)].copy()

    pub_pred_parts  = []
    pri_pred_parts  = []

    for group_name, model in group_models.items():
        mask_pub = df_test_public_all['sku_group'] == group_name
        mask_pri = df_test_private_all['sku_group'] == group_name

        df_pub_g = df_test_public_all[mask_pub].copy()
        df_pri_g = df_test_private_all[mask_pri].copy()

        if df_pub_g.empty and df_pri_g.empty:
            continue

        feature_names = model.model_public.booster_.feature_name()
        item_code_idx = feature_names.index(item_col)
        categories_goc = model.model_public.booster_.pandas_categorical[item_code_idx]

        df_pub_g[item_col] = pd.Categorical(df_pub_g[item_col], categories=categories_goc)
        df_pri_g[item_col] = pd.Categorical(df_pri_g[item_col], categories=categories_goc)

        # Kiểm tra feature
        missing_pub = set(features_for_public)  - set(df_pub_g.columns)
        missing_pri = set(features_for_private) - set(df_pri_g.columns)
        if missing_pub: raise ValueError(f"[{group_name}] Missing public features: {missing_pub}")
        if missing_pri: raise ValueError(f"[{group_name}] Missing private features: {missing_pri}")

        preds_pub, preds_pri = model.predict_submission(
            df_test_public=df_pub_g,
            df_test_private=df_pri_g
        )

        df_pub_g[target_col] = preds_pub
        df_pri_g[target_col] = preds_pri

        pub_pred_parts.append(df_pub_g)
        pri_pred_parts.append(df_pri_g)

        print(f"[{group_name}] public={len(df_pub_g)} rows | private={len(df_pri_g)} rows ✓")

    df_test_public_final  = pd.concat(pub_pred_parts,  axis=0).reset_index(drop=True)
    df_test_private_final = pd.concat(pri_pred_parts, axis=0).reset_index(drop=True)

    # =========================
    # 11. Pivot wide format
    # =========================
    def pivot_to_wide_format(df_partition, suffix):
        df_partition = df_partition.copy()
        df_partition[item_col] = df_partition[item_col].astype(str)
        df_partition = df_partition.sort_values(by=[item_col, date_col])
        df_partition['day_idx'] = df_partition.groupby(item_col).cumcount() + 1
        df_partition['F_col']   = 'F' + df_partition['day_idx'].astype(str)

        f_cols_ordered = [f'F{i}' for i in range(1, 29)]

        df_pivot = (
            df_partition
            .pivot(index=item_col, columns='F_col', values=target_col)
            .reindex(columns=f_cols_ordered)
            .reset_index()
        )
        df_pivot['id'] = df_pivot[item_col].astype(str) + suffix
        return df_pivot[['id'] + f_cols_ordered]

    df_public_wide  = pivot_to_wide_format(df_test_public_final,  suffix='_validation')
    df_private_wide = pivot_to_wide_format(df_test_private_final, suffix='_evaluation')

    df_our_predictions = pd.concat(
        [df_public_wide, df_private_wide], axis=0
    ).reset_index(drop=True)

    # =========================
    # 12. Merge vào sample submission
    # =========================
    f_cols = [f'F{i}' for i in range(1, 29)]

    df_final_submission = (
        df_test_template[['id']]
        .merge(df_our_predictions, on='id', how='left')
    )

    df_final_submission[f_cols] = df_final_submission[f_cols].fillna(0)
    df_final_submission[f_cols] = df_final_submission[f_cols].clip(lower=0)

    assert df_final_submission.shape[0] == df_test_template.shape[0]
    assert df_final_submission[f_cols].isna().sum().sum() == 0

    print("\nTạo submission hoàn tất!")
    print(f"Validation dates : {min(validation_dates)} → {max(validation_dates)}")
    print(f"Evaluation dates : {min(evaluation_dates)} → {max(evaluation_dates)}")
    print(f"Final submission shape: {df_final_submission.shape}")

    return df_final_submission, df_master, df_test_public_final, df_test_private_final
```
